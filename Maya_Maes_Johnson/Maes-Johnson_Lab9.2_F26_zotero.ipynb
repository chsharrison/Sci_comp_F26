**W.1** Make a map of the Gulf Coast in Cartopy, and figure our how to add rivers and state boundaries. Add a marker for Baton Rouge.

```python
plt.figure(figsize=(6,6))
ax = plt.axes(projection=ccrs.RotatedPole())
ax.set_extent([-98, -80, 17.5, 31], ccrs.PlateCarree()) 
ax.plot(-91.147385, 30.471165, marker='*', color='red', markersize=15, 
        transform=ccrs.PlateCarree())
ax.annotate("Baton Rouge", xy= (-95.25, 30.25), color= "red", transform=ccrs.PlateCarree())
ax.coastlines(resolution='10m')
ax.add_feature(cfeature.RIVERS)
ax.add_feature(cfeature.OCEAN)
ax.add_feature(cfeature.STATES)
```

**W.2** Do some more work on the Chipotle data from Monday

```python
chipotle = pd.read_csv('https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv', sep='\t')
price_as_float = []
for price in chipotle["item_price"]:
    price_as_float.append(float(price[1:]))

chipotle["price_float"] = price_as_float
chipotle.head()
```

```python
# Find unique values for item name
chipotle['item_name'].unique()
```

```python
# Find the least expensive menu item:
chipotle[chipotle['price_float'] == chipotle['price_float'].min()]
```
