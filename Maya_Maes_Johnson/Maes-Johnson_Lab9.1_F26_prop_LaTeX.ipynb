**W.2** Pandas filtering and sorting practice with Chipotle data

Import the dataset from this address https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv
Assign it to a variable called chipo.
1. How many products cost more than $10.00?
2. What is the price of each item?
    1. Print a data frame with only two columns item_name and item_price
    2. Sort by the name of the item
1. What was the quantity of the most expensive item ordered?
1. How many times did someone order more than one Canned Soda?


```python
chipotle = pd.read_csv('https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv', sep='\t')

chipotle.head()
```

```python
# Convert item_price to float:

price_as_float = []
for price in chipotle["item_price"]:
    price_as_float.append(float(price[1:]))

chipotle["price_float"] = price_as_float

chipotle.head()
```

```python
# How many products cost more than $10.00?
chipotle[chipotle["price_float"] > 10.].count()
```

<span style="color:olivedrab"> There are 1130 products with a price greater than $1000. </span>

```python
# What is the price of each item?
item_name_price = chipotle[['item_name', 'price_float']].copy()
item_name_price.sort_values('item_name')
```

```python
# What was the qty of the most expensive iten ordered?
chipotle[chipotle['price_float'] == chipotle['price_float'].max()]
```

<span style="color:olivedrab"> The quantity of the most expensive item sold was 15. </span>

```python
#How many times did someone order more than one canned soda?

chipotle[(chipotle['item_name'] == 'Canned Soda') & (chipotle['quantity'] > 1)].count()
```

<span style="color:olivedrab"> There 20 instances of someone ordering more than one canned soda. </span>
