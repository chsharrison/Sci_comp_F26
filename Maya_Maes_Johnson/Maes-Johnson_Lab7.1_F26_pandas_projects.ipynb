**W.2** Make a nice map of Antarctica. Label McMurdo Station

```python
plt.figure(figsize=(8,8))
ax = plt.axes(projection=ccrs.SouthPolarStereo())
ax.set_extent([-180, 180, -90, -60], ccrs.PlateCarree()) 
ax.plot(166.671029637742, -77.84409252905074, marker='*', color='red', markersize=15, 
        transform=ccrs.PlateCarree())
ax.annotate("McMurdo Station", xy= (156.67, -77.84), color= "red", transform=ccrs.PlateCarree())
ax.coastlines(resolution='50m')
ax.gridlines(draw_labels=True)
ax.add_feature(cfeature.OCEAN, color = 'skyblue')
```
