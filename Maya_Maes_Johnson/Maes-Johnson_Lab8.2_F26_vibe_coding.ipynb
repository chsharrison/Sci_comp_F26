```python
# Recall from last class, 
world_cup_2026_complete = world_cup_2026.fillna(0)
world_cup_2026_complete[world_cup_2026_complete["position"] == "DEF"]["matches_played"].hist(alpha = 0.5, bins=5)
world_cup_2026_complete[world_cup_2026_complete["position"] == "MID"]["matches_played"].hist(bins=5)
world_cup_2026_complete[world_cup_2026_complete["position"] == "FWD"]["matches_played"].hist(bins=5)
world_cup_2026_complete[world_cup_2026_complete["position"] == "GK"]["matches_played"].hist(bins=5)

plt.title("Matches Played vs. Position")
plt.legend(["DEF", "MID", "FWD", "GK"])
plt.show()
```

```python
matches_played=[
    world_cup_2026_complete[world_cup_2026_complete["position"] == "DEF"]["matches_played"],
    world_cup_2026_complete[world_cup_2026_complete["position"] == "MID"]["matches_played"],
    world_cup_2026_complete[world_cup_2026_complete["position"] == "FWD"]["matches_played"],
    world_cup_2026_complete[world_cup_2026_complete["position"] == "GK"]["matches_played"]
]

labels = ['DEF', 'MID', 'FWD', 'GK']

fig, ax = plt.subplots()
ax.set_ylabel('matches played')
bplot = ax.boxplot(matches_played,
                   tick_labels=labels)  # will be used to label x-ticks

plt.show()
```

```python
fig, ax = plt.subplots()

labels = ['DEF', 'MID', 'FWD', 'GK']
ax.violinplot(matches_played,
                  showmeans=False,
                  facecolor=[('yellow', 0.3), ('blue', 0.3), ('red', 0.3), ('green', 0.3)],
                  showmedians=True)
ax.set_title('Matches Played')
ax.legend(labels, loc= "upper right", fontsize = 7)
```

<span style="color:pink"> 

The bar plot of matches played was only able to provide a little information. While it was possible to see that the goalkeeper position and forward position had very different distributions for matches played, it was difficult to see how all four position distributions compared with one another. Rather than adjusting the alpha value of the bar plots, we can use a boxplot. From this, we can see that the median "matches played" for the goal keeper is much smaller by comparison to the other positions which all have a similar median for "matches played." We can get an even better sense of distribution shape by using violin plots to visualize the data. From this, we see that the forward and defense positions have very similar distribution shapes.
