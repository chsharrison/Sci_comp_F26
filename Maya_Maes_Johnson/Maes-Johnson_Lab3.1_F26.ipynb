<span style="color:green"> Useful for showing changes along a function (for example, a derivative). Could also be helpful showing a third data variable on a simple line plot. (https://matplotlib.org/stable/gallery/lines_bars_and_markers/multicolored_line.html) </span>


```python
# Plot 1 - Multicolored Lines (Original)
# I wanted to try working with this one because I have had to make a similar plot while using Matlab

from matplotlib.collections import LineCollection

def colored_line_between_pts(x,y,c, ax=None, **lc_kwargs):
    """
    To paraphrase what was given in the documentation, this function was defined
    to create multiple line segments in order to very smoothly create a color 
    gradient over the graph based on values you give it using the LineCollection 
    collection from Matplotlib. But I do think this is different than smoothing 
    through interpolation...
    
    Parameters
    ----------
    x, y: arrays
    c: color values
    ax: axis object to plot the colored line on
    **lc_kwargs: Additional arguments in LineCollection

    Returns a line collection of colored lines
    """

    # Below produces warnings to let the user know if they are using a keyword argument
    # or if the length of the color values is the same as the length of the x values

    if "array" in lc_kwargs:
        warnings.warn('The provided "array" keyword argument will be overridden')
    if len(c) != len(x) -1:
        warnings.warn(  "The c argument should have a length one less than the length of x and y. "
            "If it has the same length, use the colored_line function instead.")

    # Transposes the array and reshapes so that points can be easily stacked as a 
    # N x 1 x 2 array. The documentation clarifies this as 
    # (numlines) x (points per line) x 2 (for x and y)
    
    points = np.array([x,y]).T.reshape(-1,1,2)
    segments = np.concatenate([points[:-1],points[1:]], axis=1)
    lc = LineCollection(segments, **lc_kwargs)

    # START FROM HERE
    lc.set_array(c)

    return ax.add_collection(lc)
                            

x = np.linspace(0, 3 * np.pi, 500)
y = np.sin(x)
dydx = np.cos(0.5 * (x[:-1] + x[1:])) 

fig2, ax2 = plt.subplots()
line = colored_line_between_pts(x, y, dydx, ax2, linewidth=2, cmap = "viridis")
fig2.colorbar(line, ax = ax2, label = "dydx")

ax2.set_xlim(x.min(),x.max())
ax2.set_ylim(-1.1, 1.1)
ax2.set_title("Color between points")

plt.show()
```

```python
# Plot 2 - Stackplots (Original)
# I wanted to try working with this plot because in my undergraduate research, we had to make something similar
# Also I think it does a good job using dictionaries which we just learned about in DataCamp!

import matplotlib.ticker as mticker

# data from United Nations World Population Prospects (Revision 2019)
# https://population.un.org/wpp/, license: CC BY 3.0 IGO
year = [1950, 1960, 1970, 1980, 1990, 2000, 2010, 2018]
population_by_continent = {
    'Africa': [.228, .284, .365, .477, .631, .814, 1.044, 1.275],
    'the Americas': [.340, .425, .519, .619, .727, .840, .943, 1.006],
    'Asia': [1.394, 1.686, 2.120, 2.625, 3.202, 3.714, 4.169, 4.560],
    'Europe': [.220, .253, .276, .295, .310, .303, .294, .293],
    'Oceania': [.012, .015, .019, .022, .026, .031, .036, .039],
}

figs, ax = plt.subplots() # Allows for multiple plots on the same figure
# Code below plots population by continent by years using values provided in the dictionary
    # adds labels using the keys from the dictionary
ax.stackplot(year, population_by_continent.values(),
             labels=population_by_continent.keys(),
             alpha=0.8) # alpha sets transparency
ax.legend(loc='upper left', reverse = True)
ax.set_title('World Population')
ax.set_xlabel('Years')
ax.set_ylabel('Number of People (billions)')
ax.yaxis.set_minor_locator(mticker.MultipleLocator(.2)) # fixes ticks issue that we faced when plotting years on the x axis earlier

plt.show()
```
