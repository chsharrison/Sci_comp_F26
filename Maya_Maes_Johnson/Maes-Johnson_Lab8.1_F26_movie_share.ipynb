```python
# I know nothing about soccer positions, so I want to look at the distribution of minutes played per position.

world_cup_2026_complete[world_cup_2026_complete["position"] == "DEF"]["minutes_played"].hist(alpha=0.5, bins=20)
world_cup_2026_complete[world_cup_2026_complete["position"] == "MID"]["minutes_played"].hist(alpha=0.5, bins=20)
world_cup_2026_complete[world_cup_2026_complete["position"] == "FWD"]["minutes_played"].hist(alpha=0.5, bins=20)
world_cup_2026_complete[world_cup_2026_complete["position"] == "GK"]["minutes_played"].hist(alpha=0.5, bins=20)

plt.title("Minutes Played vs. Position")
plt.legend(["DEF", "MID", "FWD", "GK"])
plt.show()
```
