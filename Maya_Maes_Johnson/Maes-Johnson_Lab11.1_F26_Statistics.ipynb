**W.1** Write a function that returns a list of the first n powers of 2 (2^0, 2^1, 2^2, 2^3...2^n).

```python
def powers_of_2(num):
    first_n_powers = []
    for i in range(num):
        power = 2 ** i
        first_n_powers.append(power)
    return first_n_powers
```

```python
num = int(input("What power would you like to raise 2 to? "))
print(powers_of_2(num))
```

**W.2** Using the function above, plot the first 100 powers of 2

```python
x = np.arange(0,100,1)
y = powers_of_2(100)

plt.loglog(x,y)
plt.title("LogLog Plot of Powers of 2")
plt.xlabel("x")
plt.ylabel("x$^{2}$")
```
