**L.2** Loop through numbers from 1 to 30.<br>
If a number is divisible by 3, print "Fizz".<br>
If a number is divisible by 5, print "Buzz".<br>
If it's divisible by both, print "FizzBuzz".

```python
num_array = range(1,31)
for num in num_array:
    if (num % 3 == 0) and (num % 5 == 0): print('FizzBuzz')
    elif (num % 3 == 0): print('Fizz')
    elif (num % 5 == 0): print('Buzz')
    else: print(num)
```

```python
# Another way:
for n in range(1,31):
    if n % 3 == 0 and not(n % 5 == 0):
        print('Fizz')
    elif not(n % 3 == 0) and n % 5 == 0:
        print('Buzz')
    elif n % 3 == 0 and n % 5 == 0:
        print('FizzBuzz')
    else:
        print(n)
```

**E.2** What is the difference between np.linspace and np.arange? Explain and show how you would use them to plot $y=x^3$ from -10 to 10

<span style="color:lime"> np.linspace is similar to np.arange in that the first two parameters it accepts are the start and end values. However, rather than np.arange, np.linspace is inclusive in its bounds and its third parameter accepts the total number of values for the array. </span>

```python
# Using linspace:
x = np.linspace(-10, 10, 50)
y = x**3
plt.plot(x,y,color='lime')
```

```python
# Using arange:
x = np.arange(-10,10.25,0.25)
y = x**3
plt.plot(x,y,color='lime')
```

```python
# Visualize Random Walks
all_walks = []
for i in range(5):
    random_walk = [0]
    for x in range(100):
        step = random_walk[-1]
        dice = np.random.randint(1,7)
        if dice <= 2:
            step = max(0, step - 1)
        elif dice <= 5:
            step = step + 1 
        else:
            step = step + np.random.randint(1,7)
        random_walk.append(step)
    all_walks.append(random_walk)

np_aw = np.array(all_walks)


plt.clf

np_aw_t = np.transpose(np_aw)

plt.plot(np_aw_t)
plt.show()
```
