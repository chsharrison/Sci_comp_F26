```python
data = np.loadtxt('http://scipy-lectures.org/_downloads/populations.txt')
```

```python
# Exercise 1: Compute and print the mean and std of the populations of each species for the years in the period.
years = data[:,0]
hares = data[:,1]
lynx = data[:,2]
carrots = data[:,3]

# Hares:
print('Hares:')
print(hares.mean())
print(hares.std())

# Lynx
print('Lynx:')
print(lynx.mean())
print(lynx.std())

# Carrots
print('Carrots:')
print(carrots.mean())
print(carrots.std())
```

```python
# Exercise 2: Compute and print which year each species had the largest population.

# Hares:
max_pop_h = hares.max()
hares_max_yr = years[hares == max_pop_h]
print(f'The max population of hares was in {hares_max_yr}')

# Lynx:
max_pop_l = lynx.max()
lynx_max_yr = years[lynx == max_pop_l]
print(f'The max population of lynx was in {lynx_max_yr}')

# Carrots:
max_pop_c = carrots.max()
carrots_max_yr = years[carrots == max_pop_c]
print(f'The max population of carrots was in {carrots_max_yr}')
```

```python
##################### Boolean Operators #####################
# And, or, not: boolean is 1 or 0, True or False
me = 24
sister = 20
print(me > 21 or sister > 21)

# Boolean operators with NumPy: NumPy arrays use np.logical_and(), np.logical_or(), and np.logical_not()
array1 = np.array([3,0,10,7,20])
array2 = np.array([5,6,12,4,17])
print(np.logical_and(array1 < 9,
                    array2 <9))
```

```python
##################### If, Elif, Else #####################
# If
x = 21
if x < 25:
    print("Can't drive a rental car.")

# Else
x = 25
if x < 25:
    print("Can't drive a rental car.")
else:
    print("Can drive a rental car.")

# Elif
x = 20
if x < 21:
    print("Can't drink a beer.")
elif x < 25:
    print("Can't drive a rental car.")
else:
    print("Can drive a rental car.")
```
