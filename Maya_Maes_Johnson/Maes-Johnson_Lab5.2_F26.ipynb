```python
# Now that we have all the parts for our function, we can put it all together!
def format_topo(data):
    """
    This function returns the information from an SRTM15+ topobathy DEM that is 
    necessary for reformatting into an input file for TopoTide.

    Parameters
    ----------
    data : Global bathymetry and topography at 15 Arc Sec (SRTM15+ V2.5.5)

    Returns
    -------
    topo_dim : The dimensions of the topographic data as a tuple
    topo_arr : The topographic values formatted as NumPy array of dimensions (x,y) as given by topo_dim
    max_topo : The greatest value in the dataset (float)
    min_topo : The smallest value in the dataset (float)
    
    """
    
    import numpy as np
    topo_dim = []
    topo_vals = []
    with open(data, 'r') as topo:
        for i in range(2):
            line = next(topo).strip()
            for char in line:
                if char.isdigit():
                    topo_dim.append(int(char))
        for line in (topo.readlines() [4:]):
            for x in line.split():
                topo_vals.append(float(x))
    topo.close()
    topo_dim = tuple(topo_dim[::-1])
    topo_arr = np.array(topo_vals)
    topo_arr = np.reshape(topo_arr, topo_dim)
    min_topo = np.min(topo_arr)
    max_topo = np.max(topo_arr)
    return topo_dim, topo_arr, max_topo, min_topo
```

```python
# Put together the plotting function:

def plot_topo_as_pcolor(topo_arr, topo_dim, xres = 500, yres = 500, color_map = 'terrain'):
    """
    This function returns the topography plotted on pseudocolor plot (pcolor plot).

    Parameters
    ----------
    topo_arr : Output array of topography from format_topo()
    topo_dim : Output tuple of dimensions from format_topo()
    xres : Resolution of topography in x-direction (float); default is 500 m for SRTM15+ at equator
    yres : Resolution of topography in y-direction (float); default is 500 m for SRTM15+ at equator
    color_map : Color map (string)

    Returns
    -------
    Pcolor plot of topographic data
    
    """

    import matplotlib.pyplot as plt
    import numpy as np
    Z = topo_arr
    x_res = xres
    y_res = yres
    cm = color_map
    
    x_ticks_label = np.linspace(0, x_res*topo_dim[1], topo_dim[1]+1)
    x_ticks_actual = np.arange(0,topo_dim[1]+1,1)
    
    y_ticks_label = np.linspace(0, y_res*topo_dim[0], topo_dim[0]+1)
    y_ticks_actual = np.arange(0, topo_dim[0]+1,1)

    plt.pcolor(Z, cmap = cm)
    plt.colorbar(label = 'Elevation (m)')
    plt.xticks(x_ticks_actual, x_ticks_label)
    plt.xlabel('Horizontal Distance (m)', fontsize = 12)
    plt.yticks(y_ticks_actual, y_ticks_label)
    plt.ylabel('Vertical Distance (m)', fontsize = 12)
    plt.title('Input Topography (visualized)', fontsize = 15)

    plt.show()
```

```python
# modify plotting function, so that the x and y ticks are not insane:
def plot_topo_as_pcolor(topo_arr, 
                        topo_dim, 
                        xres = 500, 
                        yres = 500, 
                        color_map = 'terrain', 
                        show_ticks = False, 
                        save_fig = False, 
                        output_file = ''):
    """
    This function returns the topography plotted on pseudocolor plot (pcolor plot).

    Parameters
    ----------
    topo_arr : Output array of topography from format_topo()
    topo_dim : Output tuple of dimensions from format_topo()
    xres : Resolution of topography in x-direction (float); default is 500 m for SRTM15+ at equator
    yres : Resolution of topography in y-direction (float); default is 500 m for SRTM15+ at equator
    color_map : Color map (string)
    show_ticks : Display tick marks as meters (boolean); defalt is False and is suggested for datasets larger than 10 sq km
    save_fig : Option to save figure (boolear)
    output_file : Optional file name to output to if save_fig is set to True (string)

    Returns
    -------
    Pcolor plot of topographic data
    
    """

    import matplotlib.pyplot as plt
    import numpy as np
    Z = topo_arr
    x_res = xres
    y_res = yres
    cm = color_map

    plt.pcolor(Z, cmap = cm)
    plt.colorbar(label = 'Elevation (m)')
    plt.title('Input Topography (visualized)', fontsize = 15)
    if show_ticks == True:
        x_ticks_label = np.linspace(0, x_res*topo_dim[1], topo_dim[1]+1)
        x_ticks_actual = np.arange(0,topo_dim[1]+1,1)
        plt.xticks(x_ticks_actual, x_ticks_label)
        plt.xlabel('Horizontal Distance (m)', fontsize = 12)
        y_ticks_label = np.linspace(0, y_res*topo_dim[0], topo_dim[0]+1)
        y_ticks_actual = np.arange(0, topo_dim[0]+1,1)
        plt.yticks(y_ticks_actual, y_ticks_label)
        plt.ylabel('Vertical Distance (m)', fontsize = 12)
    else:
        plt.xlabel('Horizontal Cells', fontsize = 12)
        plt.ylabel('Vertical Cells', fontsize = 12)

    if save_fig == True:
        plt.savefig(output_file, dpi = 900)

    plt.show()
```
