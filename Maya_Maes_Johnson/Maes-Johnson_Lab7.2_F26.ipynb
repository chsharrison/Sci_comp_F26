```python
def largest_of_3(var1, var2, var3):
    largest = 0
    if (var1 > var2) and (var1 > var3):
        largest = var1
    elif (var2 > var1) and (var2 > var3):
        largest = var2
    elif (var3 > var1) and (var3 > var2):
        largest = var3
    else:
        if var1 == var2:
            largest = var1
        elif var1 == var3:
            largest = var1
        else: 
            largest = var2
    return largest
```

```python
# Import Packages
import os # already in Python's standard library
from PIL import Image, ImageDraw, ImageFont # Pillow already in site packages for Anaconda
import imageio.v2 as imageio
```

```python
def create_gif_from_folder(image_folder, output_path, title_text=""):
    # Supported image extensions
    valid_extensions = ('.png', '.jpg', '.jpeg', '.bmp', '.gif')
    
    # Get all image filenames sorted alphabetically
    # 'f' is defined right here inside the list comprehension:
    filenames = sorted([
        os.path.join(image_folder, item) 
        for item in os.listdir(image_folder) 
        if item.lower().endswith(valid_extensions)
    ])
    
    if not filenames:
        print("No images found in the specified folder.")
        return

    images = []
    
    for filename in filenames:
        img = Image.open(filename).convert('RGBA')
        
        # Add a title if provided
        if title_text:
            draw = ImageDraw.Draw(img)
            try:
                font = ImageFont.load_default()
            except Exception:
                font = None
            
            draw.text((10, 10), title_text, fill=(255, 255, 255, 255), font=font)
            
        images.append(img)
    
    # Save as a GIF with 1 second per frame (duration = 1000ms)
    images[0].save(
        output_path,
        save_all=True,
        append_images=images[1:],
        duration=1000,  # 1000 milliseconds = 1 second
        loop=0          # 0 means infinite loop
    )
    print(f"GIF successfully saved to {output_path}")

# Example usage (update with your actual folder path):
# create_gif_from_folder("path/to/your/image_folder", "output.gif", title_text="My Animation Title")
```

```python
def create_gif_from_folder(image_folder, output_path, title_text="", font_size=40):
    # Supported image extensions
    valid_extensions = ('.png', '.jpg', '.jpeg', '.bmp', '.gif')
    
    raw_filenames = [
        os.path.join(image_folder, item) 
        for item in os.listdir(image_folder) 
        if item.lower().endswith(valid_extensions)
    ]
    
    # Sort naturally (output1, output2, ..., output10, output11)
    filenames = sorted(
        raw_filenames, 
        key=lambda path: [int(text) if text.isdigit() else text.lower() for text in re.split('([0-9]+)', path)]
    )
    
    if not filenames:
        print("No images found in the specified folder.")
        return

    images = []
    
    for filename in filenames:
        img = Image.open(filename).convert('RGBA')
        
        # Add a title if provided
        if title_text:
            draw = ImageDraw.Draw(img)
            
            # Try loading a clean system font (e.g., Arial), fallback to default if not found
            try:
                # On Windows, Arial is usually in C:/Windows/Fonts/
                font = ImageFont.truetype("arial.ttf", font_size)
            except IOError:
                try:
                    # Mac fallback path
                    font = ImageFont.truetype("/Library/Fonts/Arial.ttf", font_size)
                except IOError:
                    font = ImageFont.load_default()
            
            # Optional: Add a simple background box or high-contrast shadow/outline for readability
            # Drawing white text at coordinates (20, 20)
            draw.text((20, 3), title_text, fill=(0, 0, 0, 255), font=font)
            
        images.append(img)
    
    # Save as a GIF with 1 second per frame (duration = 1000ms)
    images[0].save(
        output_path,
        save_all=True,
        append_images=images[1:],
        duration=150,  # 1000 milliseconds = 1 second
        loop=0          # 0 means infinite loop
    )
    print(f"GIF successfully saved to {output_path}")
```
