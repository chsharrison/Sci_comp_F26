**W.1** Make a map of the Gulf Coast in Cartopy, and figure our how to add rivers and state boundaries. Add a marker for Baton Rouge.

```python
import cartopy.crs as ccrs   #import map styles/types
import cartopy.feature as cfeature  # features such as the ocean, coastlines rivers, etc
import matplotlib.pyplot as plt

ax = plt.axes(projection=ccrs.PlateCarree())
ax.coastlines(resolution='50m')
#ax.gridlines(draw_labels=True)
ax.add_feature(cfeature.STATES, edgecolor="gray", linewidth=0.5)
ax.add_feature(cfeature.RIVERS, edgecolor="blue", linewidth=0.5)

ax.plot(-91.147385,30.471165,  'ro', markersize=5, transform=ccrs.PlateCarree())
#ax.text(-117, 33, 'San Diego', transform=ccrs.Geodetic())
ax.set_extent([-80,-100,20,35]) # set the limits of the plot
ax.text(-90.8,30.5, 'Baton Rouge', transform=ccrs.PlateCarree())

```
