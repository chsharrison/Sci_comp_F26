**W.2** Pandas filtering and sorting practice with Chipotle data

Import the dataset from this address https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv
Assign it to a variable called chipo.
1. How many products cost more than $10.00?
2. What is the price of each item?
    1. Print a data frame with only two columns item_name and item_price
    2. Sort by the name of the item
1. What was the quantity of the most expensive item ordered?
1. How many times did someone order more than one Canned Soda?


```python
import pandas as pd
chipo = pd.read_table("https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv")
chipo.head()
```

```python
#remove $ from the item price
chipo['item_price'] = chipo['item_price'].map(lambda x: x.lstrip('$').rstrip('aAbBcC'))
chipo.head()
#turn item price into float values 
chipo['item_price'] = (chipo['item_price']).astype(float)

```

```python
# collect item price values larger than 10
more_than_ten = chipo[chipo['item_price'] >10]
#Assuming that item price is the cost of one of item name and then quanity is how many were ordered total per order?
```

```python
items = more_than_ten['item_name'].unique()
print(items)
#31 items that cost more than 10 dollars 
```

```python
parced = chipo[['item_name', 'item_price']].copy()
#parced = parced.set_index('item_name')
parced_sort = parced.sort_values('item_name')
parced_sort.head()
parced_sort.drop_duplicates(subset = ['item_name', 'item_price'])
```

```python
chipo.head()
chipo_ordered_price = chipo.sort_values('item_price', ascending = False)
chipo_ordered_price.head()
#15 orders of chips and fresh tomato salsa for 44 dollars!!
```

```python
chipo.head()
canned_soda = chipo[chipo['item_name'] == 'Canned Soda']
canned_soda.head()
canned_soda.value_counts(canned_soda['quantity']>1)
#someone ordered more than 1 canned soda 20 times 
```
