```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs


fig = plt.figure(figsize=(8, 12))

# get the path of the file. It can be found in the repo data directory.

ax = plt.axes(projection=ccrs.PlateCarree())

ax.use_sticky_edges = False
# set a margin around the data
ax.set_xmargin(0.05)
ax.set_ymargin(0.10)

# add the image. Because this image was a tif, the "origin" of the image is in the
# upper left corner
ax.coastlines(resolution='50m', color='black', linewidth=1)

# mark a known place to help us geo-locate ourselves
ax.plot(-117.1625, 32.715, 'bo', markersize=7, transform=ccrs.Geodetic())
ax.text(-117, 33, 'San Diego', transform=ccrs.Geodetic())

plt.show()
```

```python
#class cartopy.crs.Spilhaus()

plt.figure(figsize=(7, 5))
ax = plt.axes(projection=ccrs.Spilhaus(rotation=45, false_easting=0.0, false_northing=0.0, globe=None)) # this rotates the map to 203 degrees East
ax.coastlines(resolution='110m')
```

```python


fig = plt.figure(figsize=(9, 7))
ax = plt.axes(projection=ccrs.SouthPolarStereo()) # this rotates the map to 203 degrees East
ax.coastlines(resolution='50m')
ax.gridlines()
# Mcmurdo station coordinates: -77.8463° S, 166.6682° E

ax.plot(166.6682, -77.8463, marker='o', markersize=12, alpha = 0.7, transform=ccrs.PlateCarree())
ax.text(160,-77, 'McMurdo Station', transform=ccrs.PlateCarree())
ax.set_extent([-180,180,-90, -60], crs=ccrs.PlateCarree())

```
