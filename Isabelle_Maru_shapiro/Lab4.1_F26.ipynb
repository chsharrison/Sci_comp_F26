```python
data = np.loadtxt('populations.txt')
year, hares, lynxes, carrots = data.T  # trick: columns to variables

import matplotlib.pyplot as plt
plt.axes([0.2, 0.1, 0.5, 0.8]) 

plt.plot(year, hares, year, lynxes, year, carrots) 

plt.legend(('Hare', 'Lynx', 'Carrot'), loc=(1.05, 0.5)) 
```

```python
#mean of each population
print(int(np.mean(hares)))
print(int((np.std(hares))))
print(int(np.mean(carrots)))
print(int((np.std(carrots))))
print(int(np.mean(lynxes)))
print(int((np.std(lynxes))))
```

```python
#make new array populations with everything except the years 
populations = data[:,1:]

#use np.argmax to get the index of the maximum value in each column of the array
max_years = np.argmax(populations, axis=0)

#find the year at each of the indexes found using the np.argmax
print("Max. year:", year[max_years])

```

```python
#specicies with the largest population for each year 
#get the index of the maximum spcies 
max_species = np.argmax(populations, axis=1)
species = np.array(['Hare', 'Lynx', 'Carrot'])
print("Max species:")
print(year)
print(species[max_species])
```

```python
above50000 = np.any(populations > 50000, axis=1)
print(year[above50000])
np.any?
```

```python
top2 = np.argsort(populations, axis=0)[:2]
print(year[top2])
```

```python
hare_grad = np.gradient(hares, 1.0)
print("diff(Hares) vs. Lynxes correlation", np.corrcoef(hare_grad, lynxes)[0,1])
plt.plot(year, hare_grad, year, -lynxes)
```
