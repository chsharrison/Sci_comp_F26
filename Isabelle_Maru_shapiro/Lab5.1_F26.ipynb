**E.2** What is the difference between np.linspace and np.arange? Explain and show how you would use them to plot $y=x^3$ from -10 to 10

np.linspace needs a start value, an end value, and the number of points you want in between the two, so you will choose how many values are in the array. np.arange needs a start value and an end value and the amount in between each step that you want, so the number of values in the array is based on the start and end value and what you want to step to be.  

to use them to plot a graph, you could use either to create your x values to enter into your equation. 



```python
#From -10 to 10 using np.linspace:

xa = np.linspace(-10,10,21)
xa

ya = xa**3
import matplotlib.pyplot as plt 

plt.plot(xa, ya)
```

```python
#From -10 to 10 using np.arange:

xb = np.arange(-10, 11, 1)
xb

yb = xa**3

plt.plot(xb, yb)
```
