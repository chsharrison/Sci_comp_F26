### Using packages within functions

We can also have import statements within a function, but this can slow the function down if the package is not loaded already

```python
def circle_area5(radius):
    if radius <= 0:
        print ("NOOO")
        return
    import numpy as np

    return np.pi * radius ** 2


circle_area5(5)
```

Is numpy imported outside the function? Let's check

We can have the function inputs have pre-defined default values, so that if you don't pass a variable it will use the default

```python
#put in defualt radius value -- if you don't give it a radius when calling the function then it will assume you mean 3

def circle_area6(radius = 3):
    import numpy as np
    return np.pi* radius**2 


circle_area6()
```

```python
# can our function handle lists? Let's check

r = [1,2,3]

circle_area6(r)
```

```python
# can our function handle arrays?

r = [1,2,3]
import numpy as np

rnp = np.array(r)
my_areas = circle_area6(rnp)

print(my_areas)
# what type of variable did the function return?
```

```python
##first rendition: DEF: do it with one dataframe

import pandas as pd

#load data and designate Nan values 
data = pd.read_csv("KM0506_hydrothermal_Mn_geochem_data.csv", na_values = [-9999, -8888, -7777])

#function to create subset of just REY values
def subset(dataframe):

    """function takes a dataframe with some geochemistry data and pulls out any REY data and adds it to a new dataframe "REY_df" But onyl works for one dataframe. """
#List of all possible REY values 
    REY = ['Pm_ppm', 'Y_ppm', 'La_ppm', 'Ce_ppm', 'Pr_ppm', 'Nd_ppm', 'Sm_ppm', 'Eu_ppm', 'Gd_ppm', 'Tb_ppm', 'Dy_ppm', 'Ho_ppm', 'Er_ppm', 'Tm_ppm', 'Yb_ppm', 'Lu_ppm']

    #empty list that will be filled with REY values actually in dataset
    data_REY = []
    
#goes thru each column name in REY and if it's in the dataframe it adds it to the empty data_REY list 
    for col in REY:
        if col in dataframe.columns:
             data_REY.append(col)
            
#make a new dataframe with all the columns that are in the data_REY list and their values 
    REY_df = dataframe[data_REY]

#add together all values in each row and put in new column REY_SUM
    REY_df["REY_SUM"] = REY_df.sum(axis = 'columns')
    
    return REY_df

REY_df = subset(data)
```

```python
## DEF do it with two dataframes 


import pandas as pd

#load data and designate Nan values 
data1 = pd.read_csv("KM0506_hydrothermal_Mn_geochem_data.csv", na_values = [-9999, -8888, -7777])
data2 = pd.read_csv("TN403_Bulk_Geochem_Data.csv", na_values = [-9999, -8888, -7777])

#function to create subset of just REY values
def subset(dataframe1, dataframe2):

    """function takes a dataframe with some geochemistry data and pulls out any REY data and adds it to a new dataframe "REY_df" """
#List of all possible REY values 
    REY = ['Pm_ppm', 'Y_ppm', 'La_ppm', 'Ce_ppm', 'Pr_ppm', 'Nd_ppm', 'Sm_ppm', 'Eu_ppm', 'Gd_ppm', 'Tb_ppm', 'Dy_ppm', 'Ho_ppm', 'Er_ppm', 'Tm_ppm', 'Yb_ppm', 'Lu_ppm']

    #empty list that will be filled with REY values actually in dataset
    data_REY1 = []
    
#goes thru each column name in REY and if it's in the dataframe it adds it to the empty data_REY list 
    for col in REY:
        if col in dataframe1.columns:
             data_REY1.append(col)
            
#make a new dataframe with all the columns that are in the data_REY list and their values 
    REY1_df = dataframe1[data_REY1]

#add together all values in each row and put in new column REY_SUM
    REY1_df["REY_SUM"] = REY1_df.sum(axis = 'columns')

########################################################
        #empty list that will be filled with REY values actually in dataset
    data_REY2 = []
    
#goes thru each column name in REY and if it's in the dataframe it adds it to the empty data_REY list 
    for col in REY:
        if col in dataframe2.columns:
             data_REY2.append(col)
            
#make a new dataframe with all the columns that are in the data_REY list and their values 
    REY2_df = dataframe2[data_REY2]

#add together all values in each row and put in new column REY_SUM
    REY2_df["REY_SUM"] = REY2_df.sum(axis = 'columns')


    
    return REY1_df, REY2_df

REY1_df, REY2_df = subset(data1, data2)

REY1_df.head()
```

```python
#DEF make a grph with two dataframes
def make_graph(dataframe1, dataframe2):
    import matplotlib.pyplot as plt
    """ This function takes the output dataframes from Def sunbset and plots them on the same figure as a boxplot. It removed the outliers so it is just zoomed in on the box part. """
    # make a new dateframe of just REY sum values together from both dataframes and plot that
    combined = pd.concat([REY1_df['REY_SUM'], REY2_df['REY_SUM']], axis = 1)
    combined.columns = ['Samoa Hotspot', 'Escanaba Trough']
    combined.boxplot(showfliers = False, patch_artist = True, boxprops = dict(facecolor = "lightgreen"), medianprops = dict(color = "green"))
    csfont = {'fontname':'Comic Sans MS'}
    hfont = {'fontname':'Helvetica'}
    plt.ylabel("Sum of all REY (ppm)", **csfont)
    plt.title("Total Rare Earth Elements", **csfont)
    #plt.rcParams.update({'font.size': 25})
    return plt.savefig("Total Rare Earth Elements no outliers.jpeg", dpi = 300)

make_graph(REY1_df,REY2_df)

```
