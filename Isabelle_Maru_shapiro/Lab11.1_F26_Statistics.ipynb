**W.1** Write a function that returns a list of the first n powers of 2 (2^0, 2^1, 2^2, 2^3...2^n).

```python

def Powers(n):
    output = []
    
    while n >-1:
        twos = "2 ^" + str(n)
        output.append(twos)
        n = n-1
    output.reverse()
    print(output)


Powers(19)
```

```python
def Powersint(n):
    output = []
    
    while n >-1:
        twos = 2**n
        twos = int(twos)
        output.append(twos)
        n = n-1
    output.reverse()
    return(output)


Powersint(19)
```
