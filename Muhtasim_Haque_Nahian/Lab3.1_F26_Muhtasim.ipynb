## 3.1.2 More plotting

Last time we made the following plot

```python
# Define the data

start = -2*np.pi
stop = 2*np.pi
#step = np.pi/8
step = np.pi/32 #smaller step

x_2 = np.arange(start,stop+step,step)
print(x_2)
```

```python
# plot the data

plt.plot(x_2, np.sin(x_2))
plt.xlabel('x')
plt.ylabel('y')
plt.title('y = sin(x)')
```

This uses an implict plotting style, instead of an object oriented approach where the elements of the plot are defined at the beginning and then can be manipulated

What we did last week was quick and dirty, and works great for quick plots. Sometimes you want more control. Don't worry about this for now, just know it exists in case you need it later or you see it somewhere.

```python
# let's go over an example plot in the old matplotlib documenation gallery

#d efine some data and give it some labels
labels = ['G1', 'G2', 'G3', 'G4', 'G5']
men_means = [20, 35, 30, 35, 27]
women_means = [25, 32, 34, 20, 25]
men_std = [2, 3, 4, 1, 2]
women_std = [3, 5, 2, 3, 3]
print(labels)
print(men_means)
print(women_means)
print(men_std)
print(women_std)
```

```python
plt.bar?
```

```python
# now we are going to make a bar plot using plt.bar, but calling it a different way than we did before

width = 0.55       # the width of the bars: can also be len(x) sequence

fig, ax = plt.subplots() # defining a figure

# plot the data first set of data
ax.bar(labels, men_means, width, yerr = men_std, label = 'Men')
```

```python
width = 0.55       # the width of the bars: can also be len(x) sequence

fig, ax = plt.subplots() # defining a figure

# plot the data first set of data
ax.bar(labels, men_means, width, yerr = men_std, label = 'Men')
# now add the women
ax.bar(labels, women_means, width, yerr = women_std, bottom = men_means, label = "Women")

# note this makes a stacked bar plot
```

```python
# now let's add a y-label

width = 0.55       
fig, ax = plt.subplots()
ax.bar(labels, men_means, width, yerr = men_std, label = 'Men')
ax.bar(labels, women_means, width, yerr = women_std, bottom = men_means, label = "Women")

# note these both do the same thing:
#plt.ylabel('Scores')
ax.set_ylabel('Scores')
```

```python
# add a title and a legend

width = 0.55       
fig, ax = plt.subplots()
ax.bar(labels, men_means, width, yerr = men_std, label = 'Men')
ax.bar(labels, women_means, width, yerr = women_std, bottom = men_means, label = "Women")

# note these both do the same thing:
plt.ylabel('Scores')
#ax.set_ylabel('Scores')

ax.set_title('Scores by group and gender')
ax.legend()
```

```python
# let's make a plot hares vs. years
from matplotlib.ticker import MaxNLocator

plt.plot(years, hares, color = 'green')

plt.xlabel('years', fontweight = 'bold')
plt.ylabel('hares', fontweight = 'bold')
plt.gca().xaxis.set_major_locator(MaxNLocator(integer=True))
```

```python
# let's plot the rest of the data and add a legend
from matplotlib.ticker import MaxNLocator

plt.plot(years, hares, color = 'green', label = 'Hares')
plt.plot(years, lynx, color = 'blue', label = 'Lynx')
plt.plot(years, carrot, color = 'orange', label = 'Carrot')

plt.xlabel('years', fontweight = 'bold')
plt.ylabel('population of hares', fontweight = 'bold')
plt.title('population plot', fontweight = 'bold')
plt.gca().xaxis.set_major_locator(MaxNLocator(integer=True))
plt.legend()
```
