### Chapter 3: 

#### Choosing a distribution

| Distribution | When to use it | Main parameters in SciPy |
|---|---|---|
| Normal | A continuous measurement has a roughly symmetric, bell-shaped distribution. | `loc` = mean, `scale` = standard deviation |
| Poisson | Count how many times an event occurs in a fixed interval, assuming independent events and a constant average rate. | `mu` = expected count in that interval |
| Exponential | Measure the waiting time until the next event when events occur independently at a constant average rate. | `scale` = mean waiting time |

For example, the **number of arrivals per hour** is a count, while the **time between arrivals** is a duration. If arrivals average 3 per hour, the mean waiting time is $1/3$ hour, or 20 minutes. The units used for `scale` and the requested waiting time must match.

#### Reading probability questions

| Question wording | Calculation | Meaning |
|---|---|---|
| Exactly $k$ events | `pmf(k)` | Probability of one exact count |
| At most $k$ events | `cdf(k)` | Probability of $k$ or fewer |
| More than $k$ events | `1 - cdf(k)` | Probability above $k$ |
| At least $k$ events | `1 - cdf(k - 1)` | Includes $k$ itself |
| Between $a$ and $b$, for a continuous measurement | `cdf(b) - cdf(a)` | Probability within the interval |
| What value do 25% of observations fall below? | `ppf(0.25)` | The 25th percentile, in the original units |

`cdf()` takes a **value** and returns a **probability**. `ppf()` works in the opposite direction. It takes a **probability** and returns the corresponding **value**.

#### Simulation and histograms

- `rvs(size=n)` generates $n$ random observations from a specified distribution.
- For a normal distribution, increasing the mean shifts the center. Increasing the standard deviation spreads the observations farther from the center.
- A 20% increase means multiplying the original value by `1.20`. A 30% increase means multiplying it by `1.30`.
- A histogram groups observations into intervals called **bins**. In a frequency histogram, bar height shows how many observations fall in each interval.

#### Why use a log transformation?

A log transformation can help when a relationship rises quickly at first and then levels off. On a log scale, doubling from 10 to 20 covers the same distance as doubling from 100 to 200. Plot the transformed values and recalculate the correlation to see whether a straight line describes the pattern more closely.

#### Correlation and causation

A correlation alone cannot establish cause and effect. For example, ice cream sales and swimming pool attendance may rise together because warm weather influences both. A third factor that helps explain an observed association is called a **confounding variable**.
