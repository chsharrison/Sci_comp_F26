```python
import pandas as pd

data = np.loadtxt('populations.txt')

df = pd.DataFrame(data, columns=['Year', 'Hare', 'Lynx', 'Carrot'])
df['Year'] = df['Year'].astype(int)  #so that years appear as int

print(df.to_string(index=False))
```

```python
#ex1
#mean and std
summary = pd.DataFrame({
    'Hare': [np.mean(data[:, 1]), np.std(data[:, 1])],
    'Lynx': [np.mean(data[:, 2]), np.std(data[:, 2])],
    'Carrot': [np.mean(data[:, 3]), np.std(data[:, 3])]}, index=['Mean', 'Std'])

display(summary.round(2))
```

```python
#ex2
#year of largest pop

hare_large = df.iloc[:, 1].argmax()
lynx_large = df.iloc[:, 2].argmax()
carrot_large = df.iloc[:, 3].argmax()

summary1 = pd.DataFrame({'Hare': [int(df.iloc[hare_large, 0])],'Lynx': [int(df.iloc[lynx_large, 0])],'Carrot': [int(df.iloc[carrot_large, 0])]},
                        index=['year of largest population'])

display(summary1)
```

```python
#ex3
#

species = np.array(['H', 'L', 'C'])

order = np.argsort(data[:, 1:], axis=1)
largest = species[order[:, -1]]

summary2 = pd.DataFrame({'Year': df['Year'], 'Largest species': largest})

display(summary2)
```

```python
#step-by-step approach to filter observations from a DataFrame based on boolean arrays: check who travelled alone in titanic
import pandas as pd

# load data
titanic = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv")
print(titanic)
# extract the alone column as a series
alone = titanic["alone"]

# keep only rows where alone is true
sel = titanic[alone]

# display the selected rows
display(sel)

# check that every selected row has alone = True
assert sel["alone"].all()
```

```python
# same query but in one-liner
# keep rows where alone is True
sel = titanic[titanic['alone']]

# print sel
print(sel)
```

```python
#step-by-step approach to filter observations from a DataFrame based on boolean arrays: passengers whose fare was over 100
# extract the fare column
fare = titanic['fare']

# check which fares are over 100
high_fare = fare > 100

# keep only rows where high_fare is True
expensive_tickets = titanic[high_fare]

# print the selected rows
print(expensive_tickets)
```

```python
#step-by-step approach to filter observations from a DataFrame based on boolean arrays: passengers whose fare was greater than 100 and less than 500
# extract the fare column
fare = titanic['fare']

# check which fares are between 100 and 500
between = np.logical_and(fare > 100, fare < 500)

# keep rows where both conditions are True
medium = titanic[between]

# print medium
print(medium)
```
