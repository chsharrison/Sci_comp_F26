### Chapter 3

#### Evaluating a classifier

A classifier can produce two kinds of predictions. `model.predict(X_test)` gives one **class label** per observation. `model.predict_proba(X_test)` gives a **probability for each class**. For a binary model with classes ordered `[0, 1]`, `[:, 1]` selects the probability of class `1` for every observation.

A **confusion matrix** compares the actual labels with the predicted labels. In `confusion_matrix(y_test, y_pred)`, rows represent actual classes and columns represent predicted classes.

| | Predicted 0 | Predicted 1 |
|---|---|---|
| **Actual 0** | True negative (TN) | False positive (FP) |
| **Actual 1** | False negative (FN) | True positive (TP) |

| Metric | Calculation | Question it answers |
|---|---|---|
| Accuracy | $(TP + TN) / (TP + TN + FP + FN)$ | What fraction of all predictions were correct? |
| Precision | $TP / (TP + FP)$ | Of the predicted positives, how many were actually positive? |
| Recall | $TP / (TP + FN)$ | Of the actual positives, how many did the model identify? |
| F1-score | $2PR / (P + R)$ | How well does the model balance precision $P$ and recall $R$? |
| Support | Number of actual observations in a class | How many observations contributed to that class’s results? |

`classification_report(y_test, y_pred)` summarizes these results by class. The **macro average** gives each class equal weight. The **weighted average** gives more weight to classes with more observations.


#### Logistic regression, the ROC curve, and AUC

**Logistic regression** estimates the probability of a class. A decision threshold converts that probability into a label. Lowering the threshold flags more observations as positive, which can catch more actual positives and produce more false alarms.

- **ROC curve.** `roc_curve(y_test, y_pred_probs)` returns the false positive rates, true positive rates, and thresholds. Plot the **false positive rate** on the x-axis and the **true positive rate** on the y-axis.
- **Reading a point.** A point at $(0.20, 0.80)$ means that, at that threshold, the model flags 20% of actual negatives and detects 80% of actual positives.
- **ROC AUC.** `roc_auc_score(y_test, y_pred_probs)` summarizes how well the model ranks positives above negatives. An AUC of 0.5 corresponds to random ranking, and 1.0 indicates perfect ranking.

#### Cross-validation and hyperparameter tuning

A **hyperparameter** is a model setting chosen before fitting, such as the number of neighbors in KNN. **K-fold cross-validation** splits the training data into $k$ folds. The model trains on $k-1$ folds and validates on the remaining fold, repeating until each fold has served as the validation set once.

| Tool or setting | Meaning |
|---|---|
| `GridSearchCV` | Evaluates every combination in the specified parameter grid. |
| `RandomizedSearchCV` | Samples a specified number of combinations. Useful when the search space is large. |
| `alpha` in ridge or lasso | Larger values apply a stronger penalty to the regression coefficients. |
| `C` in logistic regression | Smaller values apply stronger regularization. |
| `best_params_` | The settings that achieved the best mean validation score among those tested. |
| `best_score_` | The mean cross-validation score for those settings. |
| `search.score(X_test, y_test)` | The selected model’s score on the held-out test set. |

#### Preparing categorical features and missing data

| Task | Python pattern | What it does |
|---|---|---|
| Encode a categorical feature | `pd.get_dummies(df, columns=["category"], drop_first=True)` | Creates binary columns. With $k$ categories, it keeps $k-1$ columns and uses the first category as the reference. |
| Create a binary target | `np.where(condition, 1, 0)` | Assigns `1` when the condition is true and `0` otherwise. |
| Count missing values | `df.isna().sum().sort_values()` | Counts missing entries per column and sorts the counts from smallest to largest. |
| Drop selected incomplete rows | `df.dropna(subset=cols)` | Removes rows that have a missing value in any listed column. |
| Impute missing numeric values | `SimpleImputer(strategy="mean")` | Learns a separate mean for each feature and uses it to fill missing entries in that feature. |

#### Centering and scaling

`StandardScaler()` standardizes each feature using its training mean $\mu$ and standard deviation $\sigma$:

$$
z = \frac{x-\mu}{\sigma}
$$

A standardized value of $z=2$ means the observation is two training standard deviations above the feature’s training mean. Scaling matters for **KNN** because features with large numerical ranges can dominate distance calculations. It also matters for **ridge, lasso, and regularized logistic regression** because feature scales affect how the coefficient penalty acts.

#### Building and tuning a pipeline

A `Pipeline` applies preprocessing and prediction steps in order. For a classifier trained on numeric features with missing entries, the steps can be written as:

    steps = [
        ("imputer", SimpleImputer()),
        ("scaler", StandardScaler()),
        ("logreg", LogisticRegression())
    ]
    pipeline = Pipeline(steps)

During cross-validation, the pipeline learns the imputation means and scaling statistics from the **training portion of each fold**. It then applies those same values to the validation portion. This prevents validation data from influencing preprocessing.

To tune a model inside a pipeline, use `step_name__parameter`. For example, `logreg__C` tells `GridSearchCV` to tune `C` in the step named `logreg`.

#### Comparing model performance

| Prediction task | Models compared in the chapter | How to read the scores |
|---|---|---|
| Regression | Linear regression, ridge, and lasso | Higher $R^2$ indicates a better fit. Lower RMSE indicates smaller prediction errors, measured in the target’s units. |
| Classification | Logistic regression, KNN, and decision tree | Higher accuracy means a larger fraction of correct labels. Precision and recall show which types of errors the model makes. |

- **Compare models on the same folds.** `cross_val_score(model, X_train, y_train, cv=kf)` returns one score per fold. In a boxplot, the center line is the median score, and the box spans the middle 50% of scores.
- **Calculate cross-validated RMSE.** With `scoring="neg_mean_squared_error"`, use `np.sqrt(-scores)` to obtain one RMSE per fold. Average those RMSE values to report the mean fold RMSE.
