**E.2** What is the difference between np.linspace and np.arange? Explain and show how you would use them to plot $y=x^3$ from -10 to 10

np.linspace() specifies how many values i want, while np.arange() specifies the step between values.

from the graph below, we can easily observe how the number of points specified in linspace() affects the point distribution, while the spacing between points specified in arange() affects the point distribution.

```python
import matplotlib.pyplot as plt

# Using linspace: 201 points, including -10 and 10
x1 = np.linspace(-10, 10, 200)
y1 = x1 ** 3

# Using arange: step of 1; stop at 11 to include 10
x2 = np.arange(-10, 11, 1)
y2 = x2 ** 3

plt.plot(x1, y1, ".g", label="linspace")
plt.plot(x2, y2, "or", label="arange")

plt.xlabel("x", fontweight="bold")
plt.ylabel("y", fontweight="bold")
plt.title("y = x³", fontweight="bold")
plt.legend()
plt.grid()
plt.show()
```

```python
#say, i'm in step 50 of a bet. now i want to determine my next move to increase the winning chances

# Starting step
step = 50

# Roll the dice
dice=np.random.randint(1,7)

# Finish the control construct
if dice <= 2 :
    step = step - 1
elif dice <= 5 :
    step = step + 1
else :
    step = step + np.random.randint(1,7)

# Print out dice and step
print(dice)
print(step)
```

```python
#same code in for loop using random_walk()


# Initialize random_walk
random_walk = [0]

# Complete the ___
for x in range(100) :
    # Set step: last element in random_walk
    step=random_walk[-1]

    # Roll the dice
    dice = np.random.randint(1,7)

    # Determine next step
    if dice <= 2:
        # Replace below: use max to make sure step can't go below 0
        step = max(0,step-1)
    elif dice <= 5:
        step = step + 1
    else:
        step = step + np.random.randint(1,7)

    # append next_step to random_walk
    random_walk.append(step)

# Print random_walk
print(random_walk)
```

```python
#visualize the steps

plt.plot(random_walk)
plt.xlabel("Iteration", fontweight="bold")
plt.ylabel("Position (stair number)", fontweight="bold")
plt.title("random_walk steps", fontweight="bold")
plt.show()
```

```python
#simulate multiple walks to include the walks of all the winning chances

# Initialize and populate all_walks
all_walks = []
for i in range(5) :
    random_walk = [0]
    for x in range(100) :
        step = random_walk[-1]
        dice = np.random.randint(1,7)
        if dice <= 2:
            step = max(0, step - 1)
        elif dice <= 5:
            step = step + 1
        else:
            step = step + np.random.randint(1,7)
        random_walk.append(step)
    all_walks.append(random_walk)

# Convert all_walks to NumPy array: np_aw
np_aw=np.array(all_walks)

# Plot all_walks
plt.plot(np_aw)
plt.xlabel("Walk Index", fontweight="bold")
plt.ylabel("Position (stair number)", fontweight="bold")
plt.title("all_walks", fontweight="bold")
plt.show()

# Transpose np_aw: np_aw_t
np_aw_t=np.transpose(np_aw)

# Plot transpose to show all_walks over all recorded positions
plt.plot(np_aw_t)
plt.xlabel("Iteration", fontweight="bold")
plt.ylabel("Position (stair number)", fontweight="bold")
plt.title("all_walks over all recorded positions", fontweight="bold")
plt.show()
```

```python
# implemented few clumsiness:
          # changed the range() function so that the simulation is performed 20 times.
          # finished the if condition so that step is set to 0 if a random float is less or equal to 0.005, using np.random.rand().
# then visualized the distribution of the end points

# Simulate random walk 500 times
all_walks = []
for i in range(500) :
    random_walk = [0]
    for x in range(100) :
        step = random_walk[-1]
        dice = np.random.randint(1,7)
        if dice <= 2:
            step = max(0, step - 1)
        elif dice <= 5:
            step = step + 1
        else:
            step = step + np.random.randint(1,7)
        if np.random.rand() <= 0.001 :
            step = 0
        random_walk.append(step)
    all_walks.append(random_walk)

# Create and plot np_aw_t
np_aw_t = np.transpose(np.array(all_walks))

# Select last row from np_aw_t: ends
ends = np_aw_t[-1,:]

# Plot histogram of ends, display plot
plt.hist(ends)
plt.xlabel("Final position (stair number)", fontweight="bold")
plt.ylabel("Frequency (number of walks)", fontweight="bold")
plt.show()
```
