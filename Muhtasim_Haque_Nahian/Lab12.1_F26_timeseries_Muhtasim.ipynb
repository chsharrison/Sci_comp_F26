#### Let's do a linear fit of the whole timeseries

Linear regression is simply fitting a line to some data, assuming that the data is on a straight line with the equation y=**m**x+**b**, where x and y are the independent and dependent data, **m** is the slope, and **b** is the intercept. Linear regression finds m and b by minimizing the distance of all the data points to a line, getting the best fit, using calculus. See:

https://plainenglish.io/blog/linear-regression-for-dummies

https://en.wikipedia.org/wiki/Linear_regression


```python
from scipy import stats
```

```python
stats.linregress?
```

```python
# the syntax for this function is x,y, so time, ssta
ssta_trend_entire = stats.linregress(sst_ds.Year, sst_ds["Annual anomaly"])
```

**E.4** Do a linear fit of the SSTA data from 1940 to 1970. What trend do you get? How confident are you in this result? How does it compare to the trends we did in lecture (use quantitative data to explain).

```python
year_1940_1970 = sst_ds.Year[60:91]
ssta_1940_1970 = sst_ds["Annual anomaly"][60:91]

ssta_trend_1940_1970 = stats.linregress(year_1940_1970, ssta_1940_1970)
trend_1940_1970 = ssta_trend_1940_1970.slope * year_1940_1970 + ssta_trend_1940_1970.intercept

plt.plot(year_1940_1970, ssta_1940_1970, label="Annual anomaly")
plt.plot(year_1940_1970, trend_1940_1970, label="1940–1970 trend")
plt.xlabel("Year")
plt.ylabel("SSTA (°F)")
plt.legend()
plt.show()

ssta_trend_1940_1970
```

From 1940 through 1970, global sea surface temperature anomaly decreased by 0.0818 °F per decade. Both periods examined in lecture showed warming: 0.1001 °F per decade over the full record (1880–2015) and 0.2071 °F per decade from 1980–2015. I have less confidence in the mid-century cooling trend. Its p-value was 0.0644, and its 95% confidence interval ranged from −0.169 to +0.005 °F per decade. Because this interval includes zero, the fitted decline falls short of statistical significance at the 5% level

```python
print(f"1940–1970 trend: {ssta_trend_1940_1970.slope * 10:.4f} °F/decade")
print(f"1940–1970 correlation: {ssta_trend_1940_1970.rvalue:.3f}")
print(f"1940–1970 p-value: {ssta_trend_1940_1970.pvalue:.4f}")

print(f"\n1880–2015 trend: {ssta_trend_entire.slope * 10:.4f} °F/decade")
print(f"1880–2015 correlation: {ssta_trend_entire.rvalue:.3f}")
print(f"1880–2015 p-value: {ssta_trend_entire.pvalue:.2e}")
```

The linear fit for **1940–1970** indicates cooling at **0.0818 °F per decade**. I have limited confidence in this cooling trend because the relationship between year and SST anomaly is relatively weak (**r = −0.336**), and the **p-value of 0.0644** provides insufficient evidence for a nonzero slope at the commonly used 0.05 threshold.

In comparison, the full record from **1880–2015** shows warming at **0.1001 °F per decade**, with a strong positive correlation (**r = 0.836**) and a very small p-value (**8.80 × 10⁻³⁷**). These results give me greater confidence in the full-record warming trend. The comparison shows that a shorter period of estimated cooling can occur within a longer record of overall warming.

**E.6** Detrend from 1970 on and plot your result. Compare this to an ENSO (El Nino) index and speculate on what may have caused this interannual variability. How does ENSO affect SST and why?

```python
year_1970_on = sst_ds.Year[90:]
ssta_1970_on = sst_ds["Annual anomaly"][90:]

ssta_trend_1970_on = stats.linregress(year_1970_on, ssta_1970_on)
trend_1970_on = ssta_trend_1970_on.slope * year_1970_on + ssta_trend_1970_on.intercept

sst_detrended_1970_on = ssta_1970_on - trend_1970_on

plt.plot(year_1970_on, sst_detrended_1970_on)
plt.axhline(0, color="gray", linestyle="--")
plt.xlabel("Year")
plt.ylabel("Detrended SSTA (°F)")
plt.title("Global sea surface temperature after removing the 1970–2015 trend")
plt.show()
```

After detrending, the SST anomalies fluctuate above and below zero. Positive values represent years warmer than the fitted trend, while negative values represent cooler years. These alternating peaks and dips suggest interannual variability that may partly reflect ENSO.

El Niño could contribute to warm peaks because weaker trade winds allow warm water to spread eastward and reduce cold-water upwelling in the equatorial Pacific. La Niña could contribute to cool dips because stronger trade winds enhance upwelling. Other ocean and atmospheric processes also affect global SST ([NOAA](https://www.noaa.gov/understanding-el-nino)).


**E.7** Perform a five year moving average of your detrended data from E.6. Explain your result in relation to the periodicity of ENSO.

```python

ssta_filtered_5yr = np.full(len(sst_detrended_1970_on), np.nan)

# Average five consecutive years and assign the result to the middle year
for i in range(len(sst_detrended_1970_on) - 4):
    ssta_filtered_5yr[i + 2] = sst_detrended_1970_on[i:i + 5].mean().item()

plt.figure(figsize=(13, 6))
plt.plot(year_1970_on, sst_detrended_1970_on, color="gray",
         label="Annual detrended SST")
plt.plot(year_1970_on, ssta_filtered_5yr, color="darkblue", linewidth=2,
         label="Five-year moving average")
plt.axhline(0, color="black", linestyle="--", linewidth=0.7)
plt.xlabel("Year")
plt.ylabel("Detrended SST anomaly (°F)")
plt.title("Annual and five-year average detrended global SST, 1970–2015")
plt.legend()
plt.show()
```

The five-year moving average smooths the sharp peaks and dips in the annual detrended SST series. For example, the warm peak in 1998 is visible in the annual data, while the smoothed line is close to zero that year because it averages the entire 1996–2000 period. ENSO events recur irregularly, typically every **two to seven years** ([NOAA](https://www.noaa.gov/understanding-el-nino)). A five-year window can therefore include both El Niño warmth and La Niña cooling, which partly cancel when averaged. The annual curve shows these short-lived fluctuations more clearly, while the smoothed curve highlights warm or cool departures that persist across several years.
