```python
#nested functions: fuunctions inside a function

def three_shouts(word1, word2, word3):
    """Returns a tuple of strings
    concatenated with '!!!'."""

    def inner(word):
        """Returns a string concatenated with '!!!'."""
        return word + '!!!'

    return (inner(word1), inner(word2), inner(word3))

print(three_shouts('a', 'b', 'c'))
```

```python
# using the keyword nonlocal in nested functions to change the value of a nonlocal variable

def echo_shout(word):
    
    """Change the value of a nonlocal variable"""
    echo_word = word + word

    print(echo_word)

    def shout():
        """Alter a variable in the enclosing scope"""    
        nonlocal echo_word                # using echo_word in nonlocal scope
        echo_word = echo_word + "!!!"

    shout()

    print(echo_word)

echo_shout("hello")
```

```python
# functions with default arguments

def shout_echo(word1, echo=1, intense=False):
    """Concatenate echo copies of word1 and three
    exclamation marks at the end of the string."""

    echo_word = word1 * echo                   # use of echo argument

    if intense is True:                        # use of intense argument

        echo_word_new = echo_word.upper() + '!!!'
    else:
        echo_word_new = echo_word + '!!!'

    return echo_word_new

with_big_echo = shout_echo("Hey", echo=5, intense=True)

big_no_echo = shout_echo("Hey", echo=1, intense=False)

print(with_big_echo)
print(big_no_echo)
```

```python
# functions with variable-length arguments (*args)

def gibberish(*args):                   # * allows to pass different numbers of arguments. Python collects them into a tuple named args
    """Concatenate strings in *args together."""

    hodgepodge =""

    for word in args:                  
        hodgepodge += word              # shorthand for: hodgepodge = hodgepodge + word        

    return hodgepodge

one_word = gibberish("luke")

many_words = gibberish("luke", "leia", "han", "obi", "darth")

print(one_word)
print(many_words)
```

```python
# functions with variable-length keyword arguments (**kwargs)

def report_status(**kwargs):          # ** allows any number of keyword arguments—inputs written as name=value
    """Print out the status of a movie character."""

    print("\nBEGIN: REPORT\n")

    for keys, values in kwargs.items():

        print(keys + ": " + values)   # .items() provides the dictionary’s key–value pairs

    print("\nEND REPORT")

report_status(name="luke", affiliation="jedi", status="missing")

report_status(name="anakin", affiliation="sith lord", status="deceased")
```

```python
# creating a function that counts how many tweets are written in each language (column 'lang' notes the language) and application or service used to post each tweet

import pandas as pd

twitter = "https://assets.datacamp.com/production/repositories/463/datasets/82e9842c09ad135584521e293091c2327251121d/tweets.csv" 
df = pd.read_csv(twitter)

def count_entries(df, col_name = 'lang'):
    """Return a dictionary with counts of
    occurrences as value for each key."""

    cols_count = {}
    col = df[col_name]
    
    for entry in col:
# If the language is in langs_count, add 1
        if entry in cols_count.keys():              # .keys() asks: “Is this language already a key in the dictionary?”
            cols_count[entry] += 1
# Else add the language to langs_count, set the value to 1
        else:
            cols_count[entry] = 1

    return cols_count

result1 = count_entries(df, col_name = 'lang')
result2 = count_entries(df, col_name = 'source')

print(result1)
print(result2)
```
