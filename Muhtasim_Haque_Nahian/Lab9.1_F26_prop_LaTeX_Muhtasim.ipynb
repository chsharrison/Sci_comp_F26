**W.2** Pandas filtering and sorting practice with Chipotle data

Import the dataset from this address https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv
Assign it to a variable called chipo.
1. How many products cost more than $10.00?
2. What is the price of each item?
    1. Print a data frame with only two columns item_name and item_price
    2. Sort by the name of the item
1. What was the quantity of the most expensive item ordered?
1. How many times did someone order more than one Canned Soda?


```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
chipo = pd.read_csv(url, sep="\t")
chipo["item_price"] = chipo["item_price"].str.replace("$", "", regex=False).astype(float)

print(chipo.head())
```

```python
# 1. How many products cost more than $10.00?

print(chipo[chipo["item_price"] > 10].shape[0])
```

but item_price is the total price for the quantity in that row. For example, two Chicken Bowls are listed at 16.98. To count rows where the individual product costs more than $10:

```python
print(chipo[(chipo["item_price"] / chipo["quantity"]) > 10].shape[0])
```

```python
# 2. What is the price of each item?

item_prices = chipo[["item_name", "item_price"]].sort_values("item_name")
print(item_prices)
```

```python
# 3. What was the quantity of the most expensive item ordered?

most_expensive = chipo.sort_values("item_price", ascending=False).iloc[0]
print(most_expensive[["item_name", "item_price", "quantity"]])
```

```python
# 4. How many times did someone order more than one Canned Soda?

soda_orders = chipo[(chipo["item_name"] == "Canned Soda") & (chipo["quantity"] > 1)]
print(soda_orders.shape[0])
```
