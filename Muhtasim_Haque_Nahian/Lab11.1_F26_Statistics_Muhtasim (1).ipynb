## Chapter 1: Summary Statistics

### Mean and Median

- `np.mean(data)` — calculate the mean.
- `np.median(data)` — calculate the median.
- `data.agg(["mean", "median"])` — calculate both together.

### Distribution Shape

A histogram shows how numerical values are distributed across intervals called **bins**.

- **Symmetric distribution:** Mean and median are usually close.
- **Right-skewed distribution:** A long tail extends toward larger values; the mean is typically greater than the median.
- **Left-skewed distribution:** A long tail extends toward smaller values; the mean is typically less than the median.

### Quantiles, Quartiles, Quintiles, and Deciles

Quantiles mark positions in an ordered dataset.

| Division | Number of pieces | Probability levels, including endpoints |
|---|---:|---|
| Quartiles | 4 | 0, 0.25, 0.50, 0.75, 1 |
| Quintiles | 5 | 0, 0.20, 0.40, 0.60, 0.80, 1 |
| Deciles | 10 | 0, 0.10, 0.20, ..., 1 |

- **Q1:** 25th percentile.
- **Q2:** 50th percentile, or median.
- **Q3:** 75th percentile.
- The endpoints `0` and `1` correspond to the minimum and maximum.

Useful syntax:
- `np.quantile(data, 0.25)` — first quartile.
- `np.quantile(data, np.linspace(0, 1, 6))` — six boundaries for quintiles.
- `np.linspace(start, stop, num)` — generate `num` equally spaced values, including both endpoints.

### Interquartile Range and Outliers

The **interquartile range (IQR)** measures the spread of the middle 50% of the data.

$$
IQR = Q_3 - Q_1
$$

The 1.5 × IQR rule identifies potential outliers:

$$
\text{Lower cutoff} = Q_1 - 1.5(IQR)
$$

$$
\text{Upper cutoff} = Q_3 + 1.5(IQR)
$$

## Chapter 2: Random Numbers and Probability

### Probability from Observed Counts

Useful syntax:
- `.value_counts()` — count observations in each category.
- `len(df)` or `df.shape[0]` — number of rows.
- `counts / counts.sum()` — convert counts to probabilities when the counts cover all observations.

### Random Sampling

- **Without replacement:** A selected observation cannot be selected again in the same sample.
- **With replacement:** A selected observation remains available and may appear more than once.

Useful syntax:
- `df.sample(5, replace=False)` — sample five rows without replacement.
- `df.sample(5, replace=True)` — sample five rows with replacement.

Sampling without replacement changes what remains available for later draws. Sampling with replacement allows independent draws when selection probabilities stay fixed.


### Expected Value

The expected value is the long-run average of a random variable.

$$
E(X) = \sum_x xP(X=x)
$$

Multiply each possible value by its probability, then sum the products.

**Example:** If group sizes `2`, `3`, and `4` have probabilities `0.25`, `0.50`, and `0.25`:

$$
E(X) = 2(0.25) + 3(0.50) + 4(0.25) = 3
$$

An expected value can be a decimal even when individual outcomes are whole numbers.

### Continuous Uniform Distribution

A continuous uniform distribution models values that are equally likely over equal-length intervals between a minimum $a$ and maximum $b$.

For an interval entirely within those limits:

$$
P(c \leq X \leq d) = \frac{d-c}{b-a}
$$


For a continuous distribution, the probability of any single exact value is zero, so including or excluding an interval endpoint does not change its probability.

Useful syntax:
- `from scipy.stats import uniform`
- `uniform.cdf(x, loc=a, scale=b-a)` — probability of a value at or below `x`.
- `uniform.rvs(loc=a, scale=b-a, size=1000)` — simulate 1,000 values.

**Remember:** `scale` is the interval width, not the maximum value.

### Cumulative Distribution Function

The cumulative distribution function (CDF) gives:

$$
F(x) = P(X \leq x)
$$

Useful probability rules:

$$
P(X > x) = 1-F(x)
$$

$$
P(a < X \leq b) = F(b)-F(a)
$$

For continuous variables, these interval probabilities are unchanged by including or excluding the endpoints.

### Binomial Distribution

The binomial distribution models the number of successes in a fixed number of trials when:

- Each trial has two outcomes: success or failure.
- The number of trials, $n$, is fixed.
- The probability of success, $p$, stays constant.
- Trials are independent.

$$
X \sim \text{Binomial}(n,p)
$$


The probability of exactly $k$ successes is:

$$
P(X=k) = \binom{n}{k}p^k(1-p)^{n-k}
$$

Useful syntax:
- `from scipy.stats import binom`
- `binom.pmf(k, n, p)` — probability of exactly `k` successes.
- `binom.cdf(k, n, p)` — probability of `k` or fewer successes.
- `1 - binom.cdf(k, n, p)` — probability of more than `k` successes.
- `binom.rvs(n, p, size=m)` — simulate `m` sets of `n` trials.

| Question | Expression | Probability |
|---|---|---:|
| Win all 3 deals | `binom.pmf(3, 3, 0.3)` | 0.027 |
| Win 1 or fewer | `binom.cdf(1, 3, 0.3)` | 0.784 |
| Win more than 1 | `1 - binom.cdf(1, 3, 0.3)` | 0.216 |
