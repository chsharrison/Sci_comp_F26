**W.1** Write a function that returns a list of the first n powers of 2 (2^0, 2^1, 2^2, 2^3...2^n).

```python
def get_powers_of_two(n):
    return [2**i for i in range(n + 1)]
get_powers_of_two(18)
```

**W.2** Using the function above, plot the first 100 powers of 2

```python
powers = get_powers_of_two(99)
plt.plot(range(100), powers)
plt.xlabel("Exponent (n)")
plt.ylabel("2 to the power of n")
plt.title("First 100 Powers of 2")
plt.yscale("log")   # log scale: each power of 2 becomes a straight line
plt.show()
```

**Measures of center:** where the "middle" of the data is.
- **Mean:** the average. Sensitive to outliers (one huge value pulls it up).
- **Median:** the middle value when sorted. Not affected much by outliers, so better for skewed data.
- **Mode:** the most common value. Useful for categorical data.

```python
import statistics

data = [2, 4, 4, 5, 7, 9, 30]
statistics.mode(data)  # Returns 4
```

- Measure of spread tells how spread out the data is
- Variance is basically squared distance from the mean whereas std is sq.root of the variance
- `ddof=1` means "this is a sample, not the whole population".

```python
np.var(data, ddof=1)   # 93.24 -> variance
np.std(data, ddof=1)   # 9.66  -> standard deviation
```

```python
print(np.var(data, ddof=1))   
print(np.std(data, ddof=1))  
```

Quantiles is the splitting of the data into equal parts where 50% quartile is actually the median

Probability of an event = number of ways it can happen / total number of outcomes.

Sampling: picking random rows from a table.

- Normal distribution is bell-shaped and symmetric; Needs mean and std.
- Central limit theorem (CLT) is when I take many samples and calculate each sample's mean and those means form a normal distribution, even if the original data is not normal.
- Poisson distribution: number of events in a fixed time (e.g. emails per hour), with average λ (lambda).
- Exponential distribution: time between Poisson events.

- Correlation (r): strength of a straight-line relationship, from −1 to 1. 0 = no linear relationship.
- Correlation is not causation. A confounder is a third variable that affects both.
- If a relationship is curved, transform it first (e.g. `np.log`).
- Randomized controlled trial:** participants are randomly assigned to groups, so it can show cause and effect.
- Observational study: you can only show association.
- Longitudinal study:** same people followed over time.
- Cross-sectional: one snapshot in time.
