![](monty_python_index.jpg)

```python
title = "Monty Python"
```

```python
# How do I print out "Monty" and "Python"?

# I can use positive indexing in Python to print "Monty" and "Python" separately. Indexing in Python starts with index 0
# The syntax of Python is: variable_name[start:stop:step]

# Printing Monty 
print(title[0:5])
```

```python
# Printing Python

print(title[6:12])

# We can also use the following. The output will be the same

# print(title[6:])
```

```python
# You don't have to specify the beginning or the end index

# Printing Monty 

# variable_name[0:5] means that starts at index 0 and stops before index 5

print(title[0:5])
```

```python
# Printing Python without specifying the ending and the beginning
# variable_name[6:] means that starts at index 6 and go all the way to the ending

print(title[6:])
```

```python
# How do I print them out using negative indices?
```

```python
# Printing Monty 

# variable_name[0:-7] means that starts at index 0 but stop before index -7

print(title[0:-7])

# The following could be used as well
# print(title[:-7])
```

```python
# Printing Python

# variable_name[-6:] means start from the 6th character and then print all the way to the end

print(title[-6:])
```

Explain what the help file is saying

The help file is mentioning the syntax that should be used to use the round function. As we can see, it accepts two arguments. However, since the second argument mentions None, this means that it is optional. I am illustrating this below
