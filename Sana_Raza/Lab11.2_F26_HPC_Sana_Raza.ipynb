```python
die_rolls = pd.Series([1, 2, 3, 4, 5, 6])
sample_means = []
for i in range(1000):                                          # repeat 1000 times
    sample_means.append(np.mean(die_rolls.sample(5, replace=True)))   # mean of 5 rolls
np.mean(sample_means)      # about 3.5 -> close to the expected value of a die
```

```python
# correlation measures straight line relationship

df = pd.DataFrame({"x": [1, 2, 3, 4, 5], "y": [2, 4, 5, 4, 6]})
print(df["x"].corr(df["y"]))   # 0.85 means strong positive relationship
```
