**E.7** Find the approximate minimum value of the function in **E.6**. How does your definition of the x-value basis array affect this minimum? Demonstrate and explain.

```python
# First finding what is the index where the minimum y value is so I can find the corresponding x value as well

index_min = np.argmin(y)
print('Index where y value is minimum is',index_min)

#Now, finding the approximate minimum y-value
value_at_min_y_index = y[index_min]
print('The approximate minimum y-value at that index is', value_at_min_y_index)
```

```python
# Finding the x-value when the y-value is minimum

x_at_min_y_index = x[index_min]
print('The corresponding x-value is', x_at_min_y_index)
```

**Answer**: If I use a lot of points (10000), then I would get very precise approximation of the minimum value. In other words, it means that the more points for the x-values is being provided, the more precise is the approximation. I am illustrating this by using lesser points in the following:

```python
x_new = np.linspace(-5,5,11) # using 11 instead of 10000

# Now, doing the same that I did before 
y_new =  (x_new+4)*(x_new-3)*(x_new+1)**2

index_min_new = np.argmin(y_new)

print("Minimum value now is:", y_new[index_min_new])
print("Minimum x-value now is:", x_new[index_min_new])
```

**Answer**: As we can see here, with fewer data points, the estimation would be less precise.
