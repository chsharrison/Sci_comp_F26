Computes and print, based on the data in populations.txt…
1. The mean and std of the populations of each species for the years in the period.
2. Which year each species had the largest population.
3. Which species has the largest population for each year. (Hint: argsort & fancy indexing of np.array(['H', 'L', 'C']))

```python
# 1. The mean and std of the populations of hare for the years in the period.
    
# Calculate the mean (average)
hare_mean = np.mean(data[:,1])
# Calculate the standard deviation
hare_std = np.std(data[:,1])
print(hare_mean)
print(hare_std)
```

```python
# 2. The mean and std of the populations of lynx for the years in the period.

# Calculate the mean (average)
lynx_mean = np.mean(data[:,2])
# Calculate the standard deviation
lynx_std = np.std(data[:,2])
print(lynx_mean)
print(lynx_std)
```

```python
# 1. The mean and std of the populations of carrot for the years in the period.

# Calculate the mean (average)
carrot_mean = np.mean(data[:,3])
# Calculate the standard deviation
carrot_std = np.std(data[:,3])
print(carrot_mean)
print(carrot_std)
```

```python
# Which year hare had the largest population?
    
max_hare_population_year = np.argmax(data[:,1])
print(data[max_hare_population_year,0])
```

```python
# Which year lynx had the largest population?

max_lynx_population_year = np.argmax(data[:,2])
print(data[max_lynx_population_year,0])
```

```python
# Which year carrot had the largest population?

max_carrot_population_year = np.argmax(data[:,3])
print(data[max_carrot_population_year,0])
```

```python
# 3. Which species has the largest population for each year?
# There are multiple ways to do the following question. If I use pandas, then it would add the year itself. But, I am taking the use of the hints to solve this in Numpy

species = np.array(['Hare', 'Lynx', 'Carrot'])

# axis=1 just means the second dimension of the array
maxValues = species[np.argmax(data[:, 1:], axis=1)] # [:, 1:] just means that select all the rows but exclude the first column

# Since the years given are from 1900-1920, I have used range which would just take out the years
for year in range(1900, 1921):
    print(year, maxValues[year - 1900])
```
