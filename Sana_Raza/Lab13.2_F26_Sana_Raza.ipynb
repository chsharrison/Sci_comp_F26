**W.4 Pandas Practice con**

Compute the minimum, 25th percentile, median, 75th, and maximum of the following series:

ser = pd.Series(np.random.normal(10, 5, 25))

What kind of distribution is this?


````python
ser = pd.Series(np.random.normal(10, 5, 25))

# calculating the five-number summary
summary = ser.quantile([0, 0.25, 0.5, 0.75, 1])
summary.index = ['Minimum', '25th percentile', 'Median',
                 '75th percentile', 'Maximum']

print(summary)
````

This looks like a normal (Gaussian) distribution. This is because np.random.normal(10, 5, 25) is drawing 25 random numbers from a normal distribution with a mean that is 10 and a standard deviation that is 5.

**Why accuracy is not enough**: if only 5% of people have diabetes, a model that always says "no" is 95% accurate but useless. So we look at other metrics.

**Confusion matrix** (uses predicted labels, `y_pred`)
```python
from sklearn.metrics import confusion_matrix, classification_report
y_pred = model.predict(X_test)
confusion_matrix(y_test, y_pred)
# [[true negatives   false positives]
#  [false negatives  true positives ]]
```

**Classification report** (also uses `y_pred`)
```python
print(classification_report(y_test, y_pred))
```
- **precision** = of the ones predicted positive, how many really are (few false positives)
- **recall** = of the real positives, how many were caught (few false negatives)
- **f1-score** = balance of precision and recall

**Logistic regression**: a classifier (despite its name) that gives a probability between 0 and 1
```python
from sklearn.linear_model import LogisticRegression
logreg = LogisticRegression()
logreg.fit(X_train, y_train)
y_pred_probs = logreg.predict_proba(X_test)[:, 1]   # column 1 = probability of class 1
# .predict() says 1 when the probability is above 0.5
```

**ROC curve and AUC** (use probabilities, `y_pred_probs`)
```python
from sklearn.metrics import roc_curve, roc_auc_score
fpr, tpr, thresholds = roc_curve(y_test, y_pred_probs)
plt.plot([0, 1], [0, 1], 'k--')   # dashed line = random guessing
plt.plot(fpr, tpr)                # closer to the top-left corner = better
roc_auc_score(y_test, y_pred_probs)   # area under the curve: 0.5 = random, 1 = perfect
```
Remember: ROC/AUC need **probabilities**; confusion matrix and report need **labels**.

**Hyperparameter tuning**: hyperparameters are settings chosen *before* training (k in KNN, alpha in Ridge/Lasso, C in logistic regression)

GridSearchCV tries **every** combination:
```python
from sklearn.model_selection import GridSearchCV, KFold
kf = KFold(n_splits=5, shuffle=True, random_state=42)
param_grid = {"alpha": np.linspace(0.00001, 1, 20)}
lasso_cv = GridSearchCV(Lasso(), param_grid, cv=kf)
lasso_cv.fit(X_train, y_train)
lasso_cv.best_params_   # best setting
lasso_cv.best_score_    # best cross-validation score
```

RandomizedSearchCV tries only a **random sample** of combinations (faster):
```python
from sklearn.model_selection import RandomizedSearchCV
params = {"C": np.linspace(0.1, 1.0, 50),
          "class_weight": ["balanced", {0: 0.8, 1: 0.2}]}
logreg_cv = RandomizedSearchCV(LogisticRegression(), params, cv=kf)   # 10 random picks by default (n_iter)
logreg_cv.fit(X_train, y_train)
```
- **C** is the opposite of alpha: smaller C = stronger regularization
- `class_weight="balanced"` makes the rarer class count more
- Note: in newer scikit-learn (1.8+), `penalty="l1"/"l2"` is being replaced by `l1_ratio` (1 = l1, 0 = l2). It still works but gives a warning.
