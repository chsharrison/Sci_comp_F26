**W.1** Play around with some map projections in cartopy. 

```python
# Mollweide: an oval shape that keeps areas correct
fig, ax = plt.subplots(figsize=(8, 4), subplot_kw=dict(projection=ccrs.Mollweide()))
ax.coastlines()
ax.gridlines()
plt.title("Mollweide")
plt.show()
```

```python
# Orthographic
# I have added the latitude and longitude of Karachi here (longitude 67°E, latitude 25°N)
fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(projection=ccrs.Orthographic(67, 25)))
ax.coastlines()
ax.gridlines()
plt.title("Orthographic, centered on Karachi")
plt.show()
```

```python
# Interrupted Goode Homolosine: the world "peeled" like an orange
fig, ax = plt.subplots(figsize=(8, 4), subplot_kw=dict(projection=ccrs.InterruptedGoodeHomolosine()))
ax.coastlines()
ax.set_global()      
plt.title("Interrupted Goode Homolosine")
plt.show()
```

```python
# Spilhaus: an ocean-centered projection, where all oceans form one connected body of water
fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(projection=ccrs.Spilhaus()))
ax.coastlines()
ax.set_global()    
plt.title("Spilhaus")
plt.show()
```

**W.2** Make a nice map of Antarctica. Label McMurdo Station

```python
fig, ax = plt.subplots(figsize=(7, 7), subplot_kw=dict(projection=ccrs.SouthPolarStereo())) # SouthPolarStereo() actually looks straight down the south pole. It is the standard projection for Antartica

# -180 to 180 is actually the longitudes; -90 is the South Pole; -60 is the northern edge of the map (60°S, Antarctic Treaty boundary)

ax.set_extent([-180, 180, -90, -60], crs=ccrs.PlateCarree())

# Adding land, ocean, coastline and grid

ax.add_feature(cfeature.LAND, facecolor="white")
ax.add_feature(cfeature.OCEAN, facecolor="lightblue")
ax.coastlines()
ax.gridlines()

# Now, labeling McMurdo Station. Took the values from here: https://en.wikipedia.org/wiki/McMurdo_Station

ax.plot(166.67, -77.85, "r*", markersize=15, transform=ccrs.PlateCarree()) # adding star to mark the position of the station
ax.text(166.67, -76.5, "McMurdo Station", transform=ccrs.PlateCarree())
plt.title("Antarctica")
plt.show()
```
