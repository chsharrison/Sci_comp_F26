### Parsing for bad input

Note my above function will work even if the input is negative, which is not reasonable. So I probably want to include a statement that checks if the input is weird.

```python
def circle_area4(radius):
    # put in an if statement to handle bad inputs
 
    if radius <= 0:
        print ('Error: radius needs to be positive')
        return # the return statement here ends the function call
    return 3.14*radius**2
```

### Using packages within functions

We can also have import statements within a function, but this can slow the function down if the package is not loaded already

```python
def circle_area5(radius):
    if radius <= 0:
        print ('Error: radius needs to be positive')
        return # exit the function
    import numpy as np
    
    return np.pi*radius**2
```

```python
#let's test

circle_area5(5)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("parkinsons_data.csv")

# I am interested in finding out which Ethnicity has what genetic locus. I do not want to look at the whole two columns, but just the first 5 values
# A method called .head() is used to peek at the first 5 values of the column

print(data[["Ethnicity"]].head())
print(data[["Genetic locus"]].head())

# Here, C means caucasian. A means Asian, O means others ethnicity
```

```python
# Now, counting how many unique loci are present in each ethnicity
# My plan is to first find all the Ethnicity. In python language, that means I have to group all the ethnicity using .groupby() method. Then, since there can be many duplicate ethnicity, I have to use unique to find out the unique categories only!

def counting_loci_by_ethnicity(data):
    """Count unique genetic loci in each ethnicity category."""    # this is called as docstring
    counts = data.groupby("Ethnicity")["Genetic locus"].nunique()
    return counts

counting_loci_by_ethnicity(data)
```

```python
# The plot that I want to make is called as both Stem or Lollipop plot.

# Checking the parameters needed in this type of the plot

plt.stem?
```

```python
# Function 2: Making the lollipop graph which is just a minimal form of bar plot

def plotting_lollipop(counts):
    """Plot the unique locus counts in the lollipop chart"""
    plt.stem(counts.index, counts.values)


# Now, defining the x and y labels
    
    plt.xlabel("Ethnicity category")
    plt.ylabel("Number of unique genetic loci listed")
    plt.title("Genetic loci listed by ethnicity category")
    
# Mentioning the ticks on the x-axis is important as that is where I have plotted all the ethnicties
    
    plt.xticks(rotation=90)
    plt.show()


# Now, in order for me to make the lollipop plot, I have to give it a variable that has the function inside. Just calling the function in the plot call would not work
locus_counts = counting_loci_by_ethnicity(data)
plotting_lollipop(locus_counts)
```
