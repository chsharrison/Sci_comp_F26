**Line plot**:

This plot is used whenever you have to see the changes in the numerical values like time. Othertimes, this plot could be used to find the outliers, trends, patterns, peaks and sudden changes as well.

```python
# E.3: Line plot
# I have used the following resource

# https://matplotlib.org/stable/gallery/lines_bars_and_markers/simple_plot.html

# Creating the data

t = np.arange(0.0,2.0,0.01) # t represents the time

# Initiating sine wave equation

s = 1 + np.sin(2 * np.pi * t)

# fig just represents the entire figure

fig, ax = plt.subplots()

# ax represents the axes

ax.plot(t,s)
ax.set(xlabel='time (s)', ylabel='voltage (mV)',title='About as simple as it gets, folks')

# ax.grid() just adds the vertical and horizontal grids

ax.grid()

fig.savefig("test.png")
plt.show()
```

```python
# Modifications: Changing the color of the plot

fig, ax = plt.subplots()
ax.plot (t, s, color = 'red')
ax.set(xlabel='time (s)', ylabel='voltage (mV)',title='This is a red-colored sine wave')
ax.grid()
plt.show()
```

```python
# Modification: Adding multicolored graph

fig, ax = plt.subplots()

ax.plot(t, s, color="gray", linewidth=1)

ax.scatter(t, s, c=t, cmap="rainbow", s=20) # c here is just the color which would depend on the time

ax.set(xlabel="Time (s)",ylabel="Voltage (mV)",title="Voltage over Time with Multicolored Markers")

ax.grid()
plt.show()
```

```python
# Modification: Adding color bar along with sine wave


fig, ax = plt.subplots()

ax.plot(t, s, color="gray", linewidth=1)

points = ax.scatter(t, s, c=t, cmap="rainbow", s=20)

ax.set(xlabel="Time (s)",ylabel="Voltage (mV)",title="Voltage over Time with Multicolored Markers")

ax.grid()

# Adding the color bar
# points just connects the color bars with the colored scatter points
# orientation = vertical displays the color bar vertically
# pad defines how much gap should be between the color bar and the graph

color_bar = fig.colorbar(points,ax=ax,orientation="vertical",pad=0.03)

# color_bar just stores the color bar in a variable

color_bar.set_label("Time (s)")

plt.show()
```

**Histogram**:

This plot is used to show the frequency distribution of the continuous data by grouping them in separate bins. Further, you can only use numerical values in histogram. They are sometimes used to determine whether the data follows a normal(bell-curve) distribution or not. 

```python
# Link: https://matplotlib.org/stable/gallery/statistics/hist.html

# Histogram

import matplotlib.pyplot as plt
import numpy as np


# make data
np.random.seed(1)
x = 4 + np.random.normal(0, 1.5, 200)

# Creating the figure and the plots
fig, ax = plt.subplots()

# Creating histogram with 8 bins

ax.hist(x, bins=8, linewidth=0.5, edgecolor="white")

# Set the axis limits, ticks and the title

ax.set(xlim=(0, 8), xticks=np.arange(1, 8),ylim=(0, 56), yticks=np.linspace(0, 56, 9))

plt.show()
```

**Scatterplot**:

This type of plot is used for showing the relationship that exists between the two different numeric variables. Moreover, if the points are going up, then it means that there is a positive correlation. However, if the points are going down, that means that there is a negative relationship.

```python
# Scatter plot


# make the data
np.random.seed(3)
x = 4 + np.random.normal(0, 2, 24)
y = 4 + np.random.normal(0, 2, len(x))
# size and color:
sizes = np.random.uniform(15, 80, len(x))
colors = np.random.uniform(15, 80, len(x))

# plot
fig, ax = plt.subplots()

ax.scatter(x, y, s=sizes, c=colors, vmin=0, vmax=100)

ax.set(xlim=(0, 8), xticks=np.arange(1, 8),
       ylim=(0, 8), yticks=np.arange(1, 8))

plt.show()
```

```python
# Modification: Adding the boundary to the circles and changing the color

plt.style.use('_mpl-gallery')

# make the data
np.random.seed(3)
x = 4 + np.random.normal(0, 2, 24)
y = 4 + np.random.normal(0, 2, len(x))
# size and color:
sizes = np.random.uniform(15, 80, len(x))
colors = np.random.uniform(15, 80, len(x))

# plot
fig, ax = plt.subplots()

plt.scatter(x, y, marker='o', facecolors='none', edgecolors='red', s=100, linewidths=2)

ax.set(xlim=(0, 8), xticks=np.arange(1, 8), 
       ylim=(0, 8), yticks=np.arange(1, 8))

plt.show()
```

```python
# Modification: Adding the best-fit line
fig, ax = plt.subplots()
ax.scatter(x, y, marker='o', facecolors='none', edgecolors='red', s=100, linewidths=2)

slope, intercept = np.polyfit(x,y,1)
x_line = np.linspace(x.min(), x.max(), 100)
best_fit = slope * x_line + intercept
ax.plot(x_line, best_fit, color='blue',linestyle='--',linewidth=2,label='Line of best fit')
ax.legend()
plt.show()
```

```python
 # Counting the type of  the organisms but before that making the subset

org_subset = df[['Type of organism']].dropna()
org_counts = org_subset['Type of organism'].value_counts()
org_counts

# Now, making the bar plot
fig, ax = plt.subplots(figsize=(4, 4))
plt.bar(org_counts.index, org_counts.values, edgecolor='black', color=["forestgreen", "darkorange", "mediumpurple"])
plt.xlabel('Type of organism')
plt.ylabel('Number of empirical studies')
plt.title(' Organism Groups in the Dataset')
plt.xticks(rotation=45)
plt.tight_layout(rect=[0, 0.10, 1, 1])
fig.text( 0.5, 0.02, 'Number of studies showing the organsim group.', ha='center', fontsize=9, wrap=True)
plt.show()
```

```python
# Finding out which organism group has the most artifact (Node density effect)

nde_subset = df[["Type of organism", "result?"]].dropna().copy()

nde_subset["result?"] = (nde_subset["result?"].str.strip().str.upper()
)

nde = nde_subset[nde_subset["result?"] == "NDE"]

# Now, counting NDE in the organism

nde_counts = nde["Type of organism"].value_counts()
nde_counts

fig, ax = plt.subplots(figsize=(8, 5))

# nde_counts.index shows the categories(organisms) on the x-axis
# nde_counts.values defines the height of the bars

ax.bar(nde_counts.index,nde_counts.values,color=["forestgreen", "darkorange", "mediumpurple"],edgecolor="brown")

ax.set_xlabel("Types of organism")
ax.set_ylabel("Number of NDE results")
ax.set_title("NDE Results")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Analysis: There is no bar for Fungi. This shows that only two organisms group showed Node density effect (This is an important result)
```
