**L.2** Loop through numbers from 1 to 30.<br>
If a number is divisible by 3, print "Fizz".<br>
If a number is divisible by 5, print "Buzz".<br>
If it's divisible by both, print "FizzBuzz".

```python
#loop through the numbers
for n in range(1,31):
    if (n%3)==0 and (n%5==0):
        print("FizzBuzz")
    elif n%3==0:
        print("Fizz")
    elif n%5==0:
        print("Buzz")
    
```

```python
for n in range(1,31):
    if  n%3==0 and not (n%5==0):
        print("Fizz")
    elif n%5==0 and not (n%3==0):
        print("Buzz")
    elif n%5==0 and n%3==0:
        print("FizzBuzz")
```

for same seed it gives same random numbersr

randint() means random integer.

diff between random list and random walk:

random list contains separate random numbers.
random walk uses each result to determine the next position.


    

using max to fix the threshold
For example, to make sure that a variable x never goes below 10 when you decrease it, you can use:

x = max(10, x - 1)

```python
# Simulate random walk 20 times
all_walks = []
for i in range(20) :  #this is outer loop #number of this randomwalk simulation is 20 times
    random_walk = [0]
    for x in range(100) :   #100 times dice roll for on random walk #this is inner loop
        step = random_walk[-1]  #current position,last step(to the right)
        dice = np.random.randint(1,7)
        if dice <= 2:
            step = max(0, step - 1) #maximum can go to 0 number step
        elif dice <= 5:
            step = step + 1
        else:
            step = step + np.random.randint(1,7)

        # Implement clumsiness
        if np.random.rand() <= 0.005: #chances of falling down. if it gives less than or equal to 0.005 then step become zero. and for the next loop random walk[-1] will be zero.meaning start again
            step = 0

        random_walk.append(step)
    all_walks.append(random_walk)

# Create and plot np_aw_t
np_aw_t = np.transpose(np.array(all_walks))
plt.plot(np_aw_t)
plt.show()
```
