**W.2** Pandas filtering and sorting practice with Chipotle data

Import the dataset from this address https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv
Assign it to a variable called chipo.
1. How many products cost more than $10.00?
2. What is the price of each item?
    1. Print a data frame with only two columns item_name and item_price
    2. Sort by the name of the item
1. What was the quantity of the most expensive item ordered?
1. How many times did someone order more than one Canned Soda?


```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
chipo = pd.read_csv(url, sep='\t')
chipo.head()
```

```python
# Clean the price column
chipo['item_price'] = chipo['item_price'].str.replace('$', '').str.strip().astype(float)

more_than_10 = chipo[chipo['item_price'] > 10.0]
print(len(more_than_10))  
```

```python
twocolumn=chipo[['item_name', 'item_price']]
print(twocolumn)
```

```python
#sort by item name
twocolumn_sorted = chipo[['item_name', 'item_price']].sort_values(by='item_name')
print(twocolumn_sorted)
```

```python
most_expensive_quantity = chipo.loc[chipo['item_price'].idxmax()]['quantity']

print(most_expensive_quantity) 
```

```python
more_soda = chipo[(chipo['item_name'] == 'Canned Soda') & (chipo['quantity'] > 1)]

# 2. Count the number of rows that match this condition
count = len(more_soda)
print(count)
```
