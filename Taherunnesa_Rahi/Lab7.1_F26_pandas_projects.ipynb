**W.2** Make a nice map of Antarctica. Label McMurdo Station

```python
fig,ax= plt.subplots(figsize =(8,10),subplot_kw=dict(projection=ccrs.SouthPolarStereo()))
land_50m = cfeature.NaturalEarthFeature('physical', 'land', '50m',
                                    edgecolor='black',
                                        facecolor='0.9')
x=166.67
y=-77.85
ax.set_extent([-180,180,-90,-60],crs=ccrs.PlateCarree())
ax.plot(x, y,marker='*',color="red", transform=ccrs.PlateCarree())

ax.coastlines(resolution='110m')
# label the station with an offset
ax.text(166.67, -76.5, "McMurdo Station", transform=ccrs.PlateCarree())
ax.gridlines()
plt.show()
```
