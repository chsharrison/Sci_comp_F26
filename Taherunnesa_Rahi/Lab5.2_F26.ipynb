### Parsing for bad input

Note my above function will work even if the input is negative, which is not reasonable. So I probably want to include a statement that checks if the input is weird.

```python
def circle_area4(radius):
    # put in an if statement to handle bad inputs
    if radius <= 0:
        print('Error: radius needs to be positive')
        return #the return statement here ends the function call
    return 3.14*radius**2
```

```python
area3 = circle_area4(-5)
print(area3)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
bluegill= pd.read_csv("bluegill_python.csv")
bluegill
```

```python
def analyze_fish(df,column):
    """it returns the maximum, minimum, and mean of measured fish"""
    max_fish=df[column].max()
    min_fish=df[column].min()
    mean_fish=df[column].mean()
    return(max_fish,min_fish,mean_fish)
    
max_min_mean=analyze_fish(bluegill,'Number_Measured')
print(max_min_mean) 
```

```python
def makeplotfish(df, column1, column2):
    """it gives a plot of fish measured by year"""
    plt.plot(df[column1], df[column2],color="blue")

    plt.xlabel(column1)
    plt.ylabel(column2)
    plt.title("Fish measured by year",fontsize=20)
    plt.savefig("fishmeasured.png",dpi=300)
    plt.show()

makeplotfish(bluegill,"Year","Number_Measured")
```
