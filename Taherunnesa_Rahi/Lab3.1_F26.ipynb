```python
#E.6.1
import pandas as pd

water_quality = pd.read_csv(
    'WaterQualityLacDesAllemand_1991-2025.csv'
)
water_quality.head()
```

```python
# E6.2 Create a subset of the data

do_subset = water_quality[
    ['Collection Date', 'Dissolved Oxygen', 'units']
].copy()

# Rename the units column
do_subset = do_subset.rename(
    columns={'units': 'DO Unit'}
)

# Display the complete table with a caption
do_subset.style.set_caption(
    'Table 1. Dissolved oxygen observations collected from Lac des Allemands between 1991 and 2025.'
)
```

```python
# Convert Collection Date from text to date format


# Convert Collection Date to datetime

do_subset['Collection Date'] = pd.to_datetime(
    do_subset['Collection Date']
)
# Remove rows with missing dissolved oxygen values

do_subset = do_subset.dropna(
    subset=['Dissolved Oxygen']
)
```

```python
do_subset.head()
```

```python
# Extract the month number

do_subset['Month'] = do_subset['Collection Date'].dt.month
# Define seasons based on month

def define_season(month):
    
    if month in [12, 1, 2]:
        return 'Winter'
    
    elif month in [3, 4, 5]:
        return 'Spring'
    
    elif month in [6, 7, 8]:
        return 'Summer'
    
    else:
        return 'Fall'
# Assign a season to each observation

do_subset['Season'] = do_subset['Month'].apply(define_season)
```

```python
# Define the desired season order

season_order = ['Winter', 'Spring', 'Summer', 'Fall']

do_subset['Season'] = pd.Categorical(
    do_subset['Season'],
    categories=season_order,
    ordered=True
)
```

```python
# Calculate count, mean, median, and standard deviation

seasonal_summary = do_subset.groupby(
    'Season',
    observed=False
)['Dissolved Oxygen'].agg(
    Observations='count',
    Mean='mean',
    Median='median',
    Standard_Deviation='std'
)

# Round values to two decimal places

seasonal_summary = seasonal_summary.round(2)

seasonal_summary
```

```python
# Store the number of observations in each season

season_counts = do_subset.groupby(
    'Season',
    observed=False
)['Dissolved Oxygen'].count()

season_counts
```

```python
# Create a list of dissolved oxygen values for each season

seasonal_do = [
    do_subset.loc[
        do_subset['Season'] == season,
        'Dissolved Oxygen'
    ]
    for season in season_order
]
```

```python
# Create the seasonal dissolved oxygen box plot

import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(11, 7))

boxplot = ax.boxplot(
    seasonal_do,
    tick_labels=season_order,
    showmeans=True,
    patch_artist=True,
    
    # Format the mean symbols
    meanprops={
        'marker': '^',
        'markerfacecolor': 'black',
        'markeredgecolor': 'black',
        'markersize': 10
    },
    
    # Format the median lines
    medianprops={
        'color': 'black',
        'linewidth': 2
    },
    
    # Format the boxes
    boxprops={
        'facecolor': 'white',
        'edgecolor': 'black',
        'linewidth': 1.5
    },
    
    # Format the whiskers
    whiskerprops={
        'color': 'black',
        'linewidth': 1.5
    },
    
    # Format the caps
    capprops={
        'color': 'black',
        'linewidth': 1.5
    }
)

# Add the 5 mg/L threshold line
ax.axhline(
    y=5,
    color='red',
    linestyle='--',
    linewidth=2,
    label='5 mg/L threshold'
)

# Add the 2 mg/L threshold line
ax.axhline(
    y=2,
    color='darkred',
    linestyle='-.',
    linewidth=2,
    label='2 mg/L threshold'
)

# Add the number of observations below each box
for position, season in enumerate(season_order, start=1):
    
    number = season_counts[season]
    
    ax.text(
        position,
        0.25,
        f'n={number}',
        horizontalalignment='center',
        fontsize=11
    )

# Add axis labels and title
ax.set_xlabel('Season', fontsize=13)
ax.set_ylabel('Dissolved Oxygen (mg/L)', fontsize=13)

ax.set_title(
    'Seasonal Distribution of Dissolved Oxygen in Lac des Allemands',
    fontsize=15
)

# Set the y-axis limits
ax.set_ylim(0, 13)

# Add horizontal gridlines
ax.grid(
    axis='y',
    linestyle=':',
    alpha=0.4
)

# Display the threshold labels
ax.legend()

plt.show()
```

```python


# Calculate seasonal dissolved oxygen statistics

seasonal_analysis = do_subset.groupby(
    'Season',
    observed=False
)['Dissolved Oxygen'].agg(
    Observations='count',
    Mean='mean',
    Median='median',
    Standard_Deviation='std',
    Minimum='min',
    Maximum='max',
    Below_5_mg_L=lambda values: (values < 5).sum(),
    Below_2_mg_L=lambda values: (values < 2).sum()
)

# Round the results to two decimal places
seasonal_analysis = seasonal_analysis.round(2)

seasonal_analysis
```

### Methods

The water-quality data were obtained from the Louisiana Department of Environmental Quality (LDEQ) Water Data Portal for Lac des Allemands. The dataset included dissolved oxygen observations collected between 1991 and 2025. Records without dissolved oxygen measurements were removed before the analysis.

The observations were divided into four seasons based on their collection month. Winter included December through February, spring included March through May, summer included June through August, and fall included September through November. For each season, I calculated the number of observations, mean, median, standard deviation, minimum, and maximum dissolved oxygen concentration. I also counted the measurements below 5 mg/L and 2 mg/L. A box plot was used to compare the seasonal distributions.


### Results

A total of 121 dissolved oxygen observations were included in the analysis. Winter had 35 observations, spring had 27, summer had 34, and fall had 25.

Winter had the highest mean dissolved oxygen concentration at 7.99 mg/L. Summer had the lowest mean concentration at 5.61 mg/L. The median concentrations were 8.40 mg/L in winter, 7.60 mg/L in spring, 6.55 mg/L in summer, and 7.39 mg/L in fall.

The overall values ranged from 0.90 to 12.33 mg/L. The highest value, 12.33 mg/L, occurred during winter, while the lowest value, 0.90 mg/L, occurred during summer. In total, 22 observations were below 5 mg/L, including 12 summer observations. Six observations were below 2 mg/L. These results show that dissolved oxygen was generally lower during summer and that low dissolved oxygen conditions occurred most often in that season.
