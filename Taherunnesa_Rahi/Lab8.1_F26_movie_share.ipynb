```python
data[data["player_name"] == "Lionel Andrés Messi"]
```

```python
#which team has the most red cards
team_red_cards = data.groupby("team_id")["red_cards"].sum()
top_team = team_red_cards.idxmax()
most_cards = team_red_cards.max()
print(f"Team ID {top_team} had the most red cards: {most_cards}")
```

```python
team_Y_cards = data.groupby("team_id")["yellow_cards"].sum()
top_Yteam = team_Y_cards.idxmax()
most_Ycards = team_Y_cards.max()
print(f"Team ID {top_Yteam} had the most yellow cards: {most_Ycards}")
```

```python
players = ["Lionel Andrés Messi", "Kylian mbappe", "Erling Braut Haaland"]

goal_totals = [
    data.loc[data["player_name"].str.casefold() == name.casefold(), "goals"].sum()
    for name in players
] #compares names without caring about uppercase or lowercase letters.
plt.figure(figsize=(8, 5))
bars = plt.bar(players, goal_totals, color=["royalblue", "purple", "seagreen"])
plt.title("Goals by Player")
plt.ylabel("Total Goals")
plt.bar_label(bars)

plt.tight_layout()
plt.show()
```
