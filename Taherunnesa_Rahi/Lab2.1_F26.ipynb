**E.4** What is a method? How is it different from a function? Demonstrate

In Python, methods are functions that belong to a class or object. The main difference between a method and a function is that methods are called on objects or classes, while functions can exist on their own and are not tied to any specific object. Also, a method is written after the object with a period. For example, `list1.count(5)` counts how many times the number 5 appears in `list1`. In comparison, a regular function is written before the object, such as `len(list1)`.
