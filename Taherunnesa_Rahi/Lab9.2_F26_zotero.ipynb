**W.1** Make a map of the Gulf Coast in Cartopy, and figure our how to add rivers and state boundaries. Add a marker for Baton Rouge.

```python
fig, ax = plt.subplots(
    figsize=(10, 6),
    subplot_kw=dict(projection=ccrs.PlateCarree())
)

# Set the Gulf Coast boundaries
ax.set_extent([-107, -74, 13, 40])

# Add map features
ax.add_feature(cfeature.LAND, facecolor="beige")
ax.add_feature(cfeature.OCEAN, facecolor="lightblue")
ax.coastlines()

# Add states and rivers
ax.add_feature(cfeature.STATES, linewidth=0.5)
ax.add_feature(cfeature.RIVERS, edgecolor="blue")

# Mark Baton Rouge
ax.plot(-91.19, 30.45, "ro", transform=ccrs.PlateCarree())
ax.text(-91.9, 31.5, "Baton Rouge",fontsize=11,
        bbox=dict(facecolor="white", edgecolor="none", alpha=0.8), transform=ccrs.PlateCarree(),zorder=6)

ax.set_title("Map of Gulf Coast")
plt.show()
```

**W.2** Do some more work on the Chipotle data from Monday

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
chipo = pd.read_csv(url, sep='\t')
chipo.head()
```

```python
# Clean the price column
chipo['item_price'] = chipo['item_price'].str.replace('$', '').str.strip().astype(float)
```

```python
#finding the largest order
order_total = chipo.groupby("order_id")["item_price"].sum()
print("Average spending:", round(order_total.mean(), 2))
print("Largest order:", order_total.max())
```

```python

plt.plot(order_total.index, order_total.values, color="blue")

plt.xlabel("Order ID")
plt.ylabel("Order total ($)")
plt.title("Total spending for each Chipo order")

plt.show()
```

```python
# add up the quantity for each item
item_quantity = chipo.groupby("item_name")["quantity"].sum()

# select the top five
top_five = item_quantity.sort_values(ascending=False).head(5)

print(top_five)


plt.bar(top_five.index, top_five.values, color="orange")

plt.xlabel("Item")
plt.ylabel("Total quantity ordered")
plt.title("Five most ordered Chipotle items")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()

plt.show()
```
