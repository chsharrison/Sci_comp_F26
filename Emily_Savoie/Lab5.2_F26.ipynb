To add multiple parameters to a function, just add the number of parameters to the function header when defining the function. Giving a value to a parameter in a function header sets that value as the default. This will be the value used if there is no argument given for it.

```python
# Example using raise_to_power function from DataCamp
def raise_to_power(value1, value2):
    new_value = value1**value2
    return new_value

# Addition of a default parameter to the above function
def raise_to_power(value1, value2=1): # default value2 to 1 if not provided
    new_value = value1**value2
    return new_value
```

You can make functions return multiple values using tuples. Tuples are similar to lists in that they can contain multiple values, but their values can't be modified. Tuples are constructed using parentheses. Elements in tuples can be accessed using indexing like lists. To return multple values in a tuple from a function:  
new_tuple = (value1, value2,...)

```python
# Example using DataCamp raise_both function
def raise_both(value1, value2):
    """Raise value1 to the power of value2 and vice versa."""
    new_value1=value1**value2
    new_value2=value2**value1

    new_tuple = (new_value1, new_value2)
    return new_tuple
```

```python
print(raise_both(2,5))
```

Depending on where things are called or defined determines where in the program they're stored, or the scope. The different types of scope are global, local, enclosing, and built-in. 
* **Local:** Confined within function definitions and cannot be accessed outside of it. First scope searched for vairable.
* **Global:** Declared outside the function and can be accessed anywhere in the program, including inside functions. You can define variables globally by using the keyword global followed by the variable name.
* **Enclosing:** When working with nested functions the outer, or enclosing, function has it's own scope. The keyword nonlocal used with a variable in the nested function will make it apply to the enclosing function as well. When nested functions are present the order of scope checked is: local -> enclosing -> global.
* **Built-in:** Defined within Python's namespace. This is the last scope checked when a variable is not found in the other three. 

```python
# Example of local vs global scope using square function
# Only stored locally
def square(value):
    new_val = value**2
    return new_val

square(4)
```

```python
# Attempt to call new_val
new_val
```

```python
# Addition of global keyword so that global scope is affected
def square(value):
    global new_val
    new_val = value**2
    return new_val

square(5)
```

```python
# Attempt to call new_val
new_val
```

```python
# Nested function syntax
def outer(...):
    """ ... """
    x = ...

    def inner(...):
        """ ... """
        y = ...
    return ...
```

```python
# Example local vs enclosing scope using echo_shout from DataCamp exercise
# echo_shout without nonlocal keyword
def echo_shout(word):
    echo_word  = word + ' ' + word
    print(echo_word)

    def shout():
        # will look within the inner function for the variable
        echo_word = echo_word + '!!!'
    shout()
    print(echo_word)

echo_shout('hello')
```

```python
# echo_shout with nonlocal keyword
def echo_shout(word):
    echo_word  = word + ' ' + word
    print(echo_word)

    def shout():
        # will now look to the enclosing function for variable
        nonlocal echo_word
        echo_word = echo_word + '!!!'
    shout()
    print(echo_word)

echo_shout('hello')
```

Aside from the ability to set default values for arguments, you can include flexible arguments using *args in the function header. Flexible arguments allow for not having a set number of arguments in the input. Using *args turns all arguments passed into a tuple called args in the function body.

```python
# add_all example from DataCamp
def add_all(*args):
    sum_all = 0
    for num in args:
        # += operator adds and assigns to the variable
        sum_all += num
    return sum_all

add_all(2)
```

```python
add_all(3,5,6,2)
```

Using **kwargs in the function header allows for adding keyword arguments, which are arguments preceded by identifiers. This turns identifier-keyword pairs into a dictionary.


If combining *args and **kwargs, the order must be: regular parameters, *args, **kwargs.

```python
# print_all example from DataCamp
def print_all(**kwargs):
    for key,value in kwargs.items():
        print(key + ': ' + value)
```

```python
print_all(name = 'dumbledore', job = 'headmaster')
```

```python
# first function
# remove samples that were empty
# 1st step: separate to community matrix
# 2nd step: sum row-wise to get sample totals
# 3rd step: append new column with row totals and remove row if total is zero 
# 4th step: remove column 
# 5th step: add columns with diversity/richness calculations
# last step: return 2 new df with cleaned data with added analysis cols
```

```python
def clean_data(df):
    """ Calculate the total abundance and Shannon Diversity Index for every sample in the dataframe. Use the total abundance to remove samples that did not contain any organisms. 
    Return the dataframe with added columns and removed samples.
    
    Parameters
    --------------------------
    df -- Dataframe with rows being samples and columns being variables. The first column of the community matrix must be 'Nemertea' 
    
    Returns
    ---------------------------
    DataFrame that was used as input with samples with a total abundance of zero removed and additional columns of total abundance and Shannon Diversity Index"""

    # Split the inputted DataFrame into 2: one DataFrame with all site and environmental info, another containing the community matrix
    try:
        comm = df.loc[ : , 'Nemertea': ]
    except:
        print('First column of community matrix must read Nemertea')

    # Caculate the total abundance for each sample
    abun = comm.sum(axis = 1)
    
    # Add the calculated abundance to the community matrix DataFrame
    comm = pd.concat([comm, abun], axis = 1)
    # Rename the added column to 'abun'
    comm.columns = [*comm.columns[:-1], 'abun']
    # Add the 'abun' column to the original DataFrame
    df = pd.concat([df,comm['abun']], axis = 1)

    # Remove samples from the original DataFrame and the community matrix that had a total abundance of zero
    comm = comm[comm['abun'] != 0]
    df = df[df['abun'] != 0]

    # Import skbio.diversity package to calculate Shannon Diversity Index
    from skbio.diversity.alpha import shannon
    # Use .apply() to apply the shannon function to the community matrix and save to new variable
    div = comm.apply(shannon, axis = 1)

    # Add Shannon Diversity Index values to the original DataFrame and rename the column to 'shannon'
    df = pd.concat([df, div],axis=1)
    df.columns = [*df.columns[:-1],'shannon']

    # Return modified verison of original DataFrame
    return df
```

```python
data = clean_data(data)
data
```

```python
# Violin plot of shannon diversity
# args for paneling?
def comm_plot(df, x, y, wrap, wrap_num, color, xlabel = '', ylabel = ''):
    """Make violin plot of variable across sample categories.
    
    Parameters
    ----------------
    df -- DataFrame with site information, season of sampling, and any parameters desirsed for plotting
    x -- Variable to use for x axis, one of the DataFrame columns
    y -- Variable to use for y axis, one of the DataFrame columns
    color -- Variable to use for color assignment, one of DataFrame columns.
    wrap -- Variable to use to make panels of FacetGrid, one of DataFrame columns
    wrap_num -- number of columns before wrapping for FacetGrid
    xlabel -- Label for the x axis. Default is blank.
    ylabel -- Label for the y axis. Default is blank.
    
    Returns
    ---------------
    Violin plot showing desired variable across chosen sample categories""" 

    # Seaborn and pyplot packages imported for plotting
    import seaborn as sns
    import matplotlib.pyplot as plt

    # Make all column headers all lowercase to prevent issues from capitalization differences with arguments
    df.columns = [col.lower() for col in df.columns]

    # Seaborn theme setting for plot. Makes background gray
    sns.set_theme(style = 'darkgrid')

    # Plot function from Seaborn to make multi paneled violin plot
    # Use .lower() on variables to make sure input is consistent with column headers
    # hue = variable to assign colors; col = variable to use for making panels
    # col_wrap = number of columns to make before making another row for the panels
    # kind = the type of plot being used; inner = what will be displayed inside violins, 'quart' shows quartiles of data
    plot = sns.catplot(x = x.lower(), y = y.lower(), hue = color, 
                       col = wrap.lower(), col_wrap = wrap_num ,data = df, 
                       kind = 'violin', inner = 'quart', palette = 'Set2',
                       width = 0.4, alpha = 0.75)

    # Capitalize first letter of x variable values to be used for tick labels
    df[x.lower()] = df[x.lower()].str.capitalize()

    # Set the x and y axis labels
    plot.set_axis_labels(xlabel, ylabel, fontsize = 14)
    # Set x tick labels to be the unique values of x variable
    plot.set_xticklabels(df[x.lower()].unique(), fontsize = 14)
    # Adjust the size of the y tick labels
    plot.tick_params(axis = 'y', labelsize = 12)
    # Set the titles for the panels
    plot.set_titles('{col_name}', size = 16)
    # Set spacing between rows of panels
    plt.subplots_adjust(hspace = 0.2)

    # Return the figure to be able to save it after
    return(plot)
```

```python
plot = comm_plot(data, 'season', 'shannon','basin', 2, 'season', ylabel = 'Shannon Diversity')
plot.savefig('shannon_violin.png', dpi = 300)
```
