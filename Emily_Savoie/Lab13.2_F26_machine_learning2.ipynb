**W.4 Pandas Practice con**

Compute the minimum, 25th percentile, median, 75th, and maximum of the following series:

ser = pd.Series(np.random.normal(10, 5, 25))

What kind of distribution is this?


np.random.normal randomly pulls numbers from a normal distribution. These numbers have a normal distribution. It is a small sample size though so it doesn't have an obvious bell curve when plotted.

```python
ser = pd.Series(np.random.normal(10,5,25))
ser
```

```python
print(np.median(ser))
print(np.max(ser))
print(np.min(ser))
print(np.quantile(ser,[0.25,0.75]))
```
