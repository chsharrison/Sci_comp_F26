Comparison operators allow for comparison of python values and provide a boolean. They can be used between numbers, variables, strings, or a combo of numbers and variables if the variable is also numeric. Comparison operators require same object types to compare, with the exception of floats and integers. 

```python
# Syntax for comparison operators

x = 8
y = 2

# Greater >, less <
x > y
x < y 

# Greater or equal >=, less or equal <=
x >= y
x <= y

# Equals ==
x == y

# Not equal !=
x != y
```

```python
# Strings look for which would come first
'dog' < 'whale'
```

When using comparison operators with an array, python performs an element-wise comparison and provides an array of the same size of corresponding booleans. 

```python
df = np.arange(1,10,1)
```

```python
# Comparison of array to some value
df < 20
```

```python
df >= 8
```

Boolean operators allow of comparison of boolean outputs of comparison operators. Boolean operators are: and (requires all conditions are True), or (requires at least one condition is True), and not (requires none of conditions are True). The not operator negates the boolean value you use it on. 

```python
x = 14

x > 6 and x < 25
```

```python
x > 10 and x < 12
```

```python
x < 13 or x < 25
```

```python
x > 15 or x < 9
```

```python
not x > 15
```

```python
not x < 25
```

Arrays have their own boolean operators. These are logical_and(), logical_or(), and logical_not(). These allow for element wise comparsion of resulting boolean arrays from comparisons. 

```python
df1 = np.random.normal(size = (1, 5))
df1
```

```python
df2 = np.random.normal(size = (1, 5))
df2
```

```python
np.logical_and(df1 < 2, df2 > 1)
```

```python
# Select values from the array based on condition
df1[np.logical_and(df1 > -1, df1 < 1)]
```

Conditional statements allow for manipulation of code based on whether certain conditions are met. Conditional statements use if, else, and elif. 
* Format for using if:  
  if condition:  
      &emsp;expression 

* Format for using else:  
if condition:  
    &emsp;expression  
else:  
    &emsp;expression

* Format for using elif:  
if condition:  
    &emsp;expression  
elif condition:  
    &emsp;expression

* Can use all three together:  
if condition:  
    &emsp;expression  
elif condition:  
    &emsp;expression  
else:  
    &emsp;expression

Python stops at the **FIRST** satisfied condition, ignoring all other conditions after that point. 

```python
area = 12.0

if(area < 10):
    print('small')
elif(area < 14):
    print('medium')
else:
    print('large')
```

```python
# Multiple variables
area = 20.3
room = 'bed'

if room == 'kit':
    print('looking around in the kitchen')
elif room == 'bed':
    print('looking around in the bedroom')
else:
    print('looking around elsewhere')

if area > 15:
    print('big place!')
elif area > 10:
    print('medium size, nice!')
else:
    print('pretty small')
```

You can use the above techniques to filter Pandas DataFrames. There are three components to this: getting the target column, making the comparison, and using the resulting boolean Series to subset the DataFrame. You can perform these three steps in one line of code:  

df[df['col_name']condition]

Because Pandas is built on NumPy you can use the logical functions to do element wise boolean operations. The format for this would be:  

df[np.logical_and(df['col']condition1, df['col'], condition2)]


This can be done using indexing or named columns. 

```python
df3 = np.round(np.random.normal(loc  = 4, scale = 1, size = (3, 4)), decimals = 2)
df3
```

```python
df3[df3[:, 1] > 4]
```

```python
df3[np.logical_and(df3[:, 2] > 4, df3[:, 3] < 5)]
```
