**W.1** Play around with some map projections in cartopy. 

```python
# Miller projection
plt.figure(figsize = (7,5))
# Shift centeral longitude
ax = plt.axes(projection=ccrs.Miller(central_longitude = -78))
ax.coastlines(resolution='110m')
# Add color to land
ax.add_feature(cfeature.LAND, facecolor = 'forestgreen')
# Add country borders
ax.add_feature(cfeature.BORDERS)
# Color oceans
ax.add_feature(cfeature.OCEAN, color = 'turquoise')
ax.gridlines(draw_labels=True, color = 'gray')
```

```python
# EckertIII projection
plt.figure(figsize = (7,5))
# Shift centeral longitude
ax = plt.axes(projection=ccrs.EckertIII(central_longitude = -78))
ax.coastlines(resolution='110m')
# Add color to land
ax.add_feature(cfeature.LAND, facecolor = 'forestgreen')
# Add country borders
ax.add_feature(cfeature.BORDERS)
# Color oceans
ax.add_feature(cfeature.OCEAN, color = 'turquoise')
ax.gridlines(draw_labels=True, color = 'gray')
```

**W.2** Make a nice map of Antarctica. Label McMurdo Station

```python
# McMurdo Station coordinates: -77.8463° S, 166.6682° E
lat = -77.8463
long = 166.6682

# SouthPole projection
plt.figure(figsize = (9,7))


# Shift centeral longitude
ax = plt.axes(projection=ccrs.SouthPolarStereo())
ax.coastlines(resolution='50m')
# Add color to land
ax.add_feature(cfeature.LAND, facecolor = 'snow')

# Color oceans
ax.add_feature(cfeature.OCEAN, color = 'lightblue')
ax.gridlines(draw_labels=True, color = 'gray')

plt.plot(long, lat, marker = 'o', color = 'mediumseagreen', markersize = 10, transform=ccrs.PlateCarree())

ax.set_extent([-180,180,-90,-60], crs=ccrs.PlateCarree())
```
