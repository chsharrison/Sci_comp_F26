**W.2** Pandas filtering and sorting practice with Chipotle data

Import the dataset from this address https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv
Assign it to a variable called chipo.
1. How many products cost more than $10.00?
2. What is the price of each item?
    1. Print a data frame with only two columns item_name and item_price
    2. Sort by the name of the item
1. What was the quantity of the most expensive item ordered?
1. How many times did someone order more than one Canned Soda?


```python
data = pd.read_table('https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv')
data
```

```python
# How many products cost more than $10?

data2 = data.drop_duplicates(subset = ['item_name', 'item_price'])
data2['item_price']= data2['item_price'].replace('[\$,]','', regex=True).astype(float)
data2.value_counts(data2['item_price']>10)
```

With item name and price combo duplicates removed, 97 items ordered were greater than $10. 

```python
# What is the price of each item? Print a data frame with only two columns item_name and item_price. Sort by the name of the item.
# Unique combos of name and price
data3 = data2[['item_name','item_price']]
data3 = data3.sort_values(by = 'item_name')
print(data3)
```

```python
# Unique combos of name and price above $10
data4 = data2[['item_name','item_price']]
data4 = data4[data4['item_price']>10]
data4 = data4.sort_values(by = 'item_name')
print(data4)
```

```python
# What was the quantity of the most expensive item ordered?
data5 = data2[data2['item_price'] == data2['item_price'].max()]
data5
```

The quantity ordered for the most expensive item was 15. 

```python
# How many times did someone order more than one Canned Soda?
data6 = data[data['item_name']=='Canned Soda']
print(data6.value_counts(data6['quantity']>1))
```

More than one canned soda was ordered 20 times.
