**W.1** Make a map of the Gulf Coast in Cartopy, and figure our how to add rivers and state boundaries. Add a marker for Baton Rouge.

```python
plt.figure(figsize=(7, 5))
ax = plt.axes(projection=ccrs.PlateCarree())
ax.set_extent([-99, -78, 22, 38]) # set the limits of the plot
ax.coastlines(resolution='50m')
ax.gridlines(draw_labels=True, color = 'grey', linestyle = '--')

# Add state borders
ax.add_feature(cfeature.STATES, edgecolor = 'black', linestyle = '-')
# Add rivers
ax.add_feature(cfeature.RIVERS, color = 'teal')
# Add color to ocean
ax.add_feature(cfeature.OCEAN, color = 'seashell')
#Add color to land
ax.add_feature(cfeature.LAND, facecolor = 'lightgrey')
# Add marker at Baton Rouge coordinates
plt.plot(-91.15, 30.47, marker = 'o', color = 'coral', markeredgecolor = 'black', markersize = 6, transform=ccrs.PlateCarree())
```

```python
data = pd.read_table('https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv')
data
```

```python
grouped_data = data.groupby('order_id').sum()
grouped_data
```

```python
items_per_order = grouped_data.groupby('quantity').count()
items_per_order.rename(columns = {'item_name':'orders'}, inplace = True)
items_per_order = items_per_order.drop(['choice_description', 'item_price'], axis = 1)


plt.plot(items_per_order, color = 'mediumseagreen', marker = 'o')
```
