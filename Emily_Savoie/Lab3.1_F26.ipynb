**E.3** Violin plot example using box plot vs violin plot comparison from documentation. Changes will be made to just the violin plot.

Violin plots are similar to box plots and show probability distributions with the ability to add markers for medians, means, quartile ranges, etc. It uses kernal density estimation to show these distributions. They are useful for data distributions across different treatments or categories. Example: 3 different temperature treatments where feeding rates were measured.

```python
# Orginal plot code from documentation

fig, axs = plt.subplots(nrows = 1, ncols = 2, figsize = (9,4))

# Fixing random state for reproducibility of random dataset
np.random.seed(19680801)

# Generate random test data
all_data = [np.random.normal(0, std, 100) for std in range (6, 10)]

# Plot violin plot
axs[0].violinplot(all_data,
                  showmeans = False,
                  showmedians = True)
axs[0].set_title('Violin plot')

# Plot box plot
axs[1].boxplot(all_data)
axs[1].set_title('Box plot')

# Adding horizontal grid lines
for ax in axs:
    ax.yaxis.grid(True)
    ax.set_xticks([y + 1 for y in range(len(all_data))], 
                  labels = ['x1', 'x2', 'x3', 'x4'])
    ax.set_xlabel('Four separate samples')
    ax.set_ylabel('Observed values')

plt.show()
```

```python
# Changing facecolor and linecolors of violin plot

fig, axs = plt.subplots(nrows = 1, ncols = 2, figsize = (9, 4))

# Violin plot
axs[0].violinplot(all_data,
                  showmeans = False,
                  showmedians = True,
                  # facecolor to define colors; used color names
                  facecolor = ['plum', 'cadetblue', 
                               'darkseagreen', 'lightcoral'],
                 # linecolor to define color for lines of violin; used hex code
                 linecolor = '#473c8b')
axs[0].set_title('Violin plot')

# Box plot
axs[1].boxplot(all_data)
axs[1].set_title('Box plot')

# Gridlines
for ax in axs:
    ax.yaxis.grid(True)
    ax.set_xticks([y + 1 for y in range(len(all_data))],
                  labels = ['x1', 'x2', 'x3', 'x4'])
    ax.set_xlabel('Four separate samples')
    ax.set_ylabel('Observed values')

plt.show()
```

```python
# Make violin plot horizontal instead of vertical; adjust the labels and grid
# lines to reflect the orientation change
fig, axs = plt.subplots(nrows = 1, ncols = 2, figsize = (9, 4))

# Violin plot
axs[0].violinplot(all_data, 
                  showmeans = False,
                  showmedians = True,
                  # orientation argument to specify horizontal
                  orientation = 'horizontal',
                  facecolor = ['plum', 'cadetblue', 
                               'darkseagreen', 'lightcoral'],
                 linecolor = '#473c8b')
axs[0].set_title('Violin plot')

# Box plot
axs[1].boxplot(all_data,
               # Changed so the axes are correct
              orientation = 'horizontal')
axs[1].set_title('Box plot')

# Grid lines
for ax in axs:
    # Changed to xaxis.grid so the grid lines follow the observed values
    ax.xaxis.grid(True)
    # Change to yticks so it corresponds to the correct spot
    ax.set_yticks([x + 1 for x in range(len(all_data))],
                  labels = ['x1', 'x2', 'x3', 'x4'])
    # Switched so labels match
    ax.set_ylabel('Four separate samples')
    ax.set_xlabel('Observed values')

plt.show()
```

**E.4** Scatter plot example from documentation.

Scatter plots are useful for looking for correlation between two variables or help with spotting clustering in data. Example: sediment grain size and species richness. 

```python
# Original documentation code

# Fixing random state for reproducibility
np.random.seed(19680801)

N = 50
x = np.random.rand(N)
y = np.random.rand(N)
colors = np.random.rand(N)
area = (30 * np.random.rand(N))**2 # 0 to 15 point radii

plt.scatter(x, y, s = area, c = colors, alpha = 0.5)
plt.show()
```

```python
# Change point fill and edge colors
# Fixing random state for reproducibility
np.random.seed(19680801)

N = 50 
x = np.random.rand(N)
y = np.random.rand(N)
colors = np.random.rand(N)
area = (30 * np.random.rand(N))**2

plt.scatter(x, y, 
            s = area, 
            c = colors, 
            # Use cmap to set and use a predefined palette
            cmap = 'Set3',
            alpha = 0.6,
            # edgecolors to define the outline color for filled shapes
            edgecolors = 'black')

plt.show()
```

```python
# Adding gridlines and background color to scatter plot
# Fixing random state for reproducibility 
np.random.seed(19680801)

N = 50
x = np.random.rand(N)
y = np.random.rand(N)
colors = np.random.rand(N)
area = (30 * np.random.rand(N))**2

plt.scatter(x, y,
            s = area,
            c = colors,
            cmap = 'Set3',
            alpha = 0.6,
            edgecolors = 'black',
            # zorder used here and in plt.grid to make sure the grid lines
            # are placed behind the points
            zorder = 2)

# plt.gca() = get current axes to change the background color; 
# set_facecolor allows for specifying color
plt.gca().set_facecolor('#fffaf0')

#plt.grid allows for adding grid lines
plt.grid(which = 'major', # indicates adding major grid lines
         axis = 'both', # indicates adding for the x and y axis
         color = '#698b69', # hex code to specify the color of the lines
         linestyle = ':', # specify dotted lines
         zorder = 1) # zorder to make sure placed behind points

plt.show()
         
```

```python
# Change the size and shape of the points
# Fixing random state for reproducibility
np.random.seed(19680801)

N = 50 
x = np.random.rand(N)
y = np.random.rand(N)
colors = np.random.rand(N)
area = (25 * np.random.rand(N))**2


plt.scatter(x, y,
            s = 100, # change from area to number to make uniform size
            c = colors,
            marker = 'p', # marker to specify different shape for points
            cmap = 'Set3',
            alpha = 0.6,
            edgecolors = 'black',
            zorder = 2)

plt.gca().set_facecolor('#fffaf0')

plt.grid(visible = True,
         which = 'major',
         axis = 'both',
         color = '#698b69',
         linestyle = ':',
         zorder = 1)

plt.show()
```

**E.5** Line plot example from documentation. 

Line graphs are useful for displaying data over time or seeing trends in data along a line. Example: annual rainfall over a period of time.

```python
# Original code from documentation

# Data for plotting
t = np.arange(0.0, 2.0, 0.01)
s = 1 + np.sin(2 * np.pi * t)

fig, ax = plt.subplots()
ax.plot(t, s)

ax.set(xlabel = 'time (s)', ylabel = 'voltage (mV)',
       title = 'About as simple as it gets, folks')
ax.grid()

plt.show()
```

```python
# Add another line using different color and linetype

# Original data for first line
t = np.arange(0.0, 2.0, 0.01)
s = 1 + np.sin(2 * np.pi * t)

# Second line data - use np.cos instead of np.sin to make cosine instead of sine wave
s2 = 1 + np.cos(2 * np.pi * t)

fig, ax = plt.subplots()
ax.plot(t, s)
# ax.plot again to add the second line with s2 and different color and line type
ax.plot(t, s2, '--', color = 'orchid')

ax.set(xlabel = 'time (s)', ylabel = 'voltage (mV)',
       title = 'About as simple as it gets, folks')
ax.grid()

plt.show()
```

```python
# Shade the area between the two lines

# Original data for first line
t = np.arange(0.0, 2.0, 0.01)
s = 1 + np.sin(2 * np.pi * t)

# Second line data 
s2 = 1 + np.cos(2 * np.pi * t)

fig, ax = plt.subplots()
ax.plot(t, s)
ax.plot(t, s2, '--', color = 'orchid')
# ax.fill_between to fill area between lines by inputting all 3 variables
# Used alpha to make filled area a little transparent
ax.fill_between(t, s, s2,
               alpha = 0.5,
               color = 'thistle')

ax.set(xlabel = 'time (s)', ylabel = 'voltage (mV)',
       title = 'About as simple as it gets, folks')
ax.grid()

plt.show()
```

```python
# Add legend and change grid lines

# Original data for first line
t = np.arange(0.0, 2.0, 0.01)
s = 1 + np.sin(2 * np.pi * t)

# Second line data
s2 = 1 + np.cos(2 * np.pi * t)

fig, ax = plt.subplots()
# Added label to both ax.plot calls to define what will be in legend
ax.plot(t, s, label = 'Sine wave')
ax.plot(t, s2, '--', color = 'orchid', label = 'Cosine wave')
ax.fill_between(t, s, s2,
                alpha = 0.5,
                color = 'thistle')

ax.set(xlabel = 'time (s)', ylabel = 'voltage (mV)',
       title = 'About as simple as it gets, folks')
# ax.legend to make legend on plot
ax.legend()
# Added color and linestyle argument to change appearance of grid lines
ax.grid(color = 'cadetblue',
       linestyle = ':')

plt.show()
```
