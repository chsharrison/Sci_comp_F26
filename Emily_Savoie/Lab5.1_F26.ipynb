**L.1** Calculate the factorial of a number.

First think about steps required to accomplish goal. For factorial, you need:  
* a number N
* to iterate down from N to 1
* multiply as you go down

It's helpful to use print statements when developing code to help with troubleshoot any issues encountered. 

```python
N = 50
prod = 1
temp = N

while temp > 0:
    prod = prod * temp
    temp = temp - 1
    #print(prod)
    #print(temp)

print(prod)
```

To figure out probabilities of something occurring, you can simulate runs in python and calculate the statistical likelihood of success. Example problem: what is the probability of reaching step number 75 of the empire state building based on the results of 150 dice throws and a small chance of falling down to the beginning. Below is a breakdown of the components used in this problem and will be built up to calculate the probability.  


First you need to be able to generate random numbers. Numpy's random() function is excellent for this. Setting the seed() allows for a reproducible set of pseudo-random numbers. To do this you just type in a number to np.random.seed(). This can be any number.

```python
# Set the seed for reproducibility
np.random.seed(12345)

# Use np.random.randint to generate a random integer between 1 and 6 to simulate a dice roll
# Range goes from first number to one less then the second number
dice = np.random.randint(1,7)
print(dice)
```

To determine the next move, or step, based on the dice throws, you need if-elif-else statements.

```python
# Random starting step for example 
step = 40

# 'Roll' the dice
dice = np.random.randint(1,7)

# Make the control statements
if dice <= 2:
    step = step - 1
elif 3 <= dice <= 5:
    step = step + 1
else:
    step = step + np.random.randint(1,7)

# Print out to check
print(dice)
print(step)
```

The next part is repeating this step process over and over to generate a succession of steps. This is known as a random walk. Use the control statements inside of a for loop to build a list containing all of the steps taken. There needs to be a statement preventing the value of step from going below zero.    

```python
# Create list to fill starting at step 0
random_walk = [0]

# Make for loop with control statements to simulate rolling dice 250 times. Add each step to list. 
for x in range(150):
    #Select the last element of random_walk to be the step value
    step = random_walk[-1]

    # Roll the dice
    dice = np.random.randint(1,7)

    # Control statements for determining next step
    # Use max() on first if statement to prevent step value from going below 0
    if dice <= 2:
        step = max(0, step - 1)
    elif dice <= 5:
        step = step + 1
    else:
        step = step + np.random.randint(1,7)

    # Add newly calculated step to the random_walk list
    random_walk.append(step)

# Print the result
print(random_walk)
```

To be able to calculate the probability of accomplishing the gaol, you need to run this multiple times and see how many of the simulations were a success. This is done be using nested for loops. The first indicates the number of times to run the simulation and the second loop is for the simulation run itself. Each simulation result will be appended to a new list.

```python
# Create empty list to add all resulting random_walk runs into 
all_walks = []

# Nested for loops to simulate run multiple times. Let's start with 10 runs
for i in range(10):
    # Code from previous step
    random_walk = [0]
    for x in range(150):
        step = random_walk[-1]
        dice = np.random.randint(1,7)

        if dice <= 2:
            step = max(0, step - 1)
        elif dice <= 5:
            step = step + 1
        else:
            step = step + np.random.randint(1,7)
        
        random_walk.append(step)
    # Add resulting random walk to all_walks
    all_walks.append(random_walk)
    
print(all_walks)
```

To account for the potential of having to start over, you need to add in another if statement to the above. To incorporate a 0.75% chance of starting over, the if statement needs to generate a random number and restart if it's below the thershold. np.random.rand() generates a random float between 0 - 1 and is excellent for this. This additional if statment needs to go within the for loop that is generating the individual runs, not the one controlling how many times to simulate it. 

```python
# Simulate 25 times
all_walks = []
for i in range(25):
    random_walk = [0]
    for x in range(150):
        step = random_walk[-1]
        dice = np.random.randint(1,7)
        if dice <= 2:
            step = max(0, step - 1)
        elif dice <= 5:
            step = step + 1
        else:
            step = step + np.random.randint(1,7)

        # Incoroporate failure rate
        if np.random.rand() <= 0.0075:
            step = 0

        random_walk.append(step)
    all_walks.append(random_walk)

# Let's plot the results
# Conver all_walks to an array
np_aw = np.array(all_walks)
# Transpose the array so that instead of the last column being the final step the last row is
np_aw_t = np.transpose(np_aw)

# Plot the transposed array. Each line is one of the runs. This doesn't give info for probablility but provides a visual for what occurred across simulations
plt.plot(np_aw_t)
plt.show()
```

Since we want to know the probability of success and success is determined by where the random walks end, you need to pull the final index of the resulting walks. A histrogram is an excellent graph for plotting distributions of the resutls seen. 

```python
# Set the seed
np.random.seed(12345)
# Simulate 600 times
all_walks = []
for i in range(600):
    random_walk = [0]
    for x in range(150):
        step = random_walk[-1]
        dice = np.random.randint(1,7)
        if dice <= 2:
            step = max(0, step - 1)
        elif dice <= 5:
            step = step + 1
        else:
            step = step + np.random.randint(1,7)
        if np.random.rand() <= 0.0075:
            step = 0

        random_walk.append(step)
    all_walks.append(random_walk)

# Create the transposed array in one step
np_aw_t = np.transpose(np.array(all_walks))

# Select the last row from all columns of the transposed array to get a list containing the final step from all the runs
ends = np_aw_t[-1, :]

# Make a histogram to show distribution
plt.hist(ends, bins=25, histtype='bar',color='lightseagreen', edgecolor = 'black')
# Add a line indicating what the 'success' thershold was, in this case 80
plt.axvline(x = 80, color = 'firebrick', linestyle = '--', label = 'success thershold')
plt.legend()
plt.show
```

To get the mathematical probability of successfully reaching step 80, we can calculate this by dividing the number of ends that are greater or equal to 80 by 600 (total number of simulations). To convert to a percent, this value should be multiplied by 100.

```python
# Assign the number of elements that are 80 or higher to a variable
success = len(ends[ends>=80])
# Divide the success variable by the total number of simulations then multiply by 100
chance = (success/600)*100
print(chance)
```
