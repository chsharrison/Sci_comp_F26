while loops are similar to if statements except it will continue to execute the command for as long as the conidition is true. The syntax for while loops is as follows:  
while condition:  
    &emsp;expression

You are able to run conditional statements inside loops. 

```python
offset = 8

while offset != 0:
    print('adjusting')
    offset = offset - 1
    print(offset)
```

```python
offset = -8

while offset !=0:
    print('adjusting')
    if offset > 0:
        offset = offset - 1
    else:
        offset = offset + 1
    print(offset)
```

for loops allow for iterative calls over a sequence. You specify the variable to use in the for loop syntax.

for var in seq:  
    &emsp;expression

When using for loops on lists, the var name is arbitrary. When a list of lists is used, the arbitrary names are assigned in the order they are listed.

```python
areas = [11.25, 18.0, 20.0, 10.75, 9.50]

for a in areas:
    print(a)
```

```python
# Use enumerate() to add the index to the output and a print statement to format it
# Adjust so index starts at 1 not 0
for index, a in enumerate(areas):
    print('room ' + str(index + 1) + ': ' + str(a))
```

```python
# Example with a list of lists
house = [["hallway", 11.25], 
         ["kitchen", 18.0], 
         ["living room", 20.0], 
         ["bedroom", 10.75], 
         ["bathroom", 9.50]]

for room, area in house:
    print('the ' + room + ' is ' + str(area) + ' sqm')
```

You can use for loops over other data structures like NumPy arrays and dictionaries. To do this, you have to adjust the sequence call. 

For disctionaries to go through each key and value, you use the items() method. The variable names are still arbitrary for dictionaries and are used in the order they are listed in the call. 

```python
europe = {'spain':'madrid', 'france':'paris', 'germany':'berlin',
          'norway':'oslo', 'italy':'rome', 'poland':'warsaw', 'austria':'vienna' }

for key, val in europe.items():
    print('the capital of ' + key + ' is ' + val)
```

1D NumPy arrays generally follow the same syntax as lists. 

2D NumPy arrays you use np.nditer(array).

```python
np_height = np.array([1.73, 1.68, 1.71, 1.89, 1.79])
np_weight = np.array([65.4, 59.2, 63.6, 88.4, 68.7])
bmi = np_weight/np_height ** 2

for val in bmi:
    print(val)
```

```python
meas = np.array([np_height, np_weight])

for val in np.nditer(meas):
    print(val)
```

```python
pop = pd.read_csv('populations.txt', sep = '\t', index_col = 0)
pop
```
