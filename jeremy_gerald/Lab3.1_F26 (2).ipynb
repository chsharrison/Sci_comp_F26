```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load the data using your specific local file path
file_path = r"C:\Users\jerem\OneDrive\Documents\21st_Century_Storm_data.xlsx"
df = pd.read_excel(file_path)

# 2. Sort chronologically
df = df.sort_values(by=['Sid', 'Yr', 'Mo', 'Da', 'hr'])

# 3. Calculate 24-hour wind speed changes for each storm
df['Wmax_24h_diff'] = df.groupby('Sid')['Wmax'].diff(periods=24)

# 4. Determine if a storm experienced RI (>= 30 knots in 24 hours)
storm_summary = df.groupby(['SYear', 'Sid'])['Wmax_24h_diff'].max().reset_index()
storm_summary['Experienced_RI'] = storm_summary['Wmax_24h_diff'] >= 30

# 5. Group by Year to get counts
yearly_counts = storm_summary.groupby('SYear').agg(
    Total_Storms=('Sid', 'count'),
    RI_Storms=('Experienced_RI', 'sum')
).reset_index()

# Calculate the mean number of RI storms per year
mean_ri_storms = yearly_counts['RI_Storms'].mean()

# 6. Create the visualization
fig, ax = plt.subplots(figsize=(12, 6))

# Bar graph for Total Storms
ax.bar(yearly_counts['SYear'], yearly_counts['Total_Storms'], 
       color='lightsteelblue', edgecolor='black', label='Total Storms')

# Line graph for RI Storms
ax.plot(yearly_counts['SYear'], yearly_counts['RI_Storms'], 
        color='crimson', marker='o', linewidth=2.5, markersize=7, label='Storms with >=1 RI Cycle')

# Add the Mean line for RI storms (Green Dashed Line)
ax.axhline(y=mean_ri_storms, color='green', linestyle='--', linewidth=2, 
           label=f'Mean RI Storms per Year ({mean_ri_storms:.1f})')

# Formatting
ax.set_xlabel('Year', fontsize=12, fontweight='bold')
ax.set_ylabel('Number of Storms', fontsize=12, fontweight='bold')
ax.set_title('Total Storms vs. Storms Experiencing Rapid Intensification (RI) per Year', fontsize=14, fontweight='bold', pad=15)
ax.grid(axis='y', linestyle='--', alpha=0.7)
ax.legend(loc='upper right', fontsize=11)

# Ensure integer ticks on the y-axis
ax.yaxis.set_major_locator(plt.MaxNLocator(integer=True))

plt.tight_layout()
plt.show()
```
