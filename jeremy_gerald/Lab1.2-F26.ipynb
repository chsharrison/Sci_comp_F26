```python
name_first = ["Luke", "Aniken", "Leia", "Owen", "Buru"]
name_last = ["Skywalker", "Skywalker", "Organa", "Lars", "Lars"] 
height = [1.72, 2.02, 1.5, 1.78, 1.65] 
weight = [77, 136, 49, 120, 75] 
age = [19, 41, 19, 52, 47]

bmi = []
bmi_cat = []

i = 0
while i < len(name_first):
    current_bmi = round(weight[i] / (height[i] ** 2), 2)
    bmi.append(current_bmi)
    

    if current_bmi < 18.5:
        bmi_cat.append("Underweight")
    elif current_bmi < 25.0:
        bmi_cat.append("Healthy Weight")
    elif current_bmi < 30.0:
        bmi_cat.append("Overweight")
    elif current_bmi < 35.0:
        bmi_cat.append("Obesity Class 1")
    elif current_bmi < 40.0:
        bmi_cat.append("Obesity Class 2")
    else:
        bmi_cat.append("Obesity Class 3 (Severe Obesity)")

    print(f"{name_first[i]} {name_last[i]}: {current_bmi}, ({bmi_cat[i]})")
    i += 1
```
