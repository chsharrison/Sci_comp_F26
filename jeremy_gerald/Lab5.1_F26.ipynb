**E.2** What is the difference between np.linspace and np.arange? Explain and show how you would use them to plot $y=x^3$ from -10 to 10

```python
import matplotlib.pyplot as plt
import numpy as np

# np.linespace (numbers are inclusive)
x_linspace = np.linspace(-10, 10, 100)
y_linspace = x_linspace**3

# np.arrange (numbers are exclusive)
x_arange = np.arange(-10, 10.1, .1)
y_arange = x_arange**3

# Plot
plt.figure(figsize=(8, 5))
plt.plot(
    x_linspace,
    y_linspace,
    label="y = x^3 (via linspace)",
    color="blue",
    linewidth=2,
)

plt.scatter(
    x_arange[::5],
    y_arange[::5],
    label="Sampled points (via arange)",
    color="red",
    zorder=5,
)

plt.title("Plot of y = x^3")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.show()
```
