```python
## E.6.1 Import the Data
import pandas as pd
btr = pd.read_csv('BTR_Aug2026.csv')

print(btr.shape)

print(btr.columns)
```

```python
btr['day'] = pd.to_datetime(btr['day'])
btr.head()
```

```python
## E.6.2 Data Table
table_data = btr[['day', 'max_temp_f', 'min_temp_f', 'precip_in']]

table_data
```

**Table 1.** Daily maximum temperature, minimum temperature, and precipitation at Baton Rouge/Ryan Airport (BTR) for the first ten days of August 2026.

```python
## E.6.3 Daily Temperature Plot
fig, ax = plt.subplots(figsize=(11, 5.5))

# Daily maximum temperature
ax.plot(
    btr['day'],
    btr['max_temp_f'],
    marker='o',
    markersize=4,
    linewidth=2,
    label='Daily Maximum'
)

# Daily minimum temperature
ax.plot(
    btr['day'],
    btr['min_temp_f'],
    marker='o',
    markersize=4,
    linewidth=2,
    label='Daily Minimum'
)

# Shade the range between daily maximum and minimum
ax.fill_between(
    btr['day'],
    btr['min_temp_f'],
    btr['max_temp_f'],
    alpha=0.12,
    label='Daily Temperature Range'
)

# Labels and title
ax.set_xlabel('Date')
ax.set_ylabel('Temperature (°F)')
ax.set_title(
    'Daily Maximum and Minimum Temperature\n'
    'Baton Rouge/Ryan Airport (BTR), August 2026'
)

# Grid
ax.grid(axis='y', linestyle='--', alpha=0.3)

# Legend
ax.legend(frameon=False)

# Plot
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.xticks(rotation=45)
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()
```

Figure 1. Daily maximum and minimum air temperatures at Baton Rouge/Ryan Airport (BTR) during August 2026. The shaded region represents the daily temperature range between the maximum and minimum temperatures.

```python
## E.6.4 Basic Analysis
max_temperature = btr['max_temp_f'].max()
min_temperature = btr['min_temp_f'].min()

mean_max_temperature = btr['max_temp_f'].mean()
mean_min_temperature = btr['min_temp_f'].mean()


total_precipitation = btr['precip_in'].sum()
```

```python
print('Highest temperature:', max_temperature, '°F')
print('Lowest temperature:', min_temperature, '°F')
print('Mean daily maximum temperature:', mean_max_temperature, '°F')
print('Mean daily minimum temperature:', mean_min_temperature, '°F')
print('Total precipitation:', total_precipitation, 'inches')
```
