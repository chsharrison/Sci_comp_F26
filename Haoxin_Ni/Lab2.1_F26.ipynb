**E.4** What is a method? How is it different from a function? Demonstrate

A method is a function that belongs to a specific object.

The main difference is how we call them, a function is called by its name with arguments like function(abc), while a method is called directly like object.method().

```python
# Function example
nums = [3, 1, 4]
print("function:", sorted(nums))

# Method example
nums.append(2)
print("method:", nums)
```
