**E.6** Plot the function $y=(x+4)(x-3)(x+1)^2$ over the interval [-5, 5]. Make sure you include the endpoints of this interval and use enough points so that your plot looks smooth. Label your axes and add a title that includes the function you just plotted. Save this plot to both a .png and .pdf file using python commands, and copy and paste the plot directly from this notebook into a word document. 

```python
import matplotlib.pyplot as plt

# Generate enough x points, 500 may be enough :O
x = np.linspace(-5, 5, 500)
y = (x + 4) * (x - 3) * ((x + 1) ** 2)

# Create the plot
plt.figure(figsize=(8, 5))
plt.plot(x, y, label=r'$y=(x+4)(x-3)(x+1)^2$', color='b')
plt.xlabel('x')
plt.ylabel('y')
plt.title(r'Plot of $y=(x+4)(x-3)(x+1)^2$')
plt.grid(True)
plt.legend()

# Save the plot to both .png and .pdf files
plt.savefig('/Users/finley/Jupyter/polynomial_plot.png', dpi=300)
plt.savefig('/Users/finley/Jupyter/polynomial_plot.pdf')

plt.show()
```

**E.7** Find the approximate minimum value of the function in **E.6**. How does your definition of the x-value basis array affect this minimum? Demonstrate and explain.

```python
# 500 points
x_fine = np.linspace(-5, 5, 500)
y_fine = (x_fine + 4) * (x_fine - 3) * ((x_fine + 1) ** 2)
approx_min_fine = np.min(y_fine)

# 25 points
x_coarse = np.linspace(-5, 5, 25)
y_coarse = (x_coarse + 4) * (x_coarse - 3) * ((x_coarse + 1) ** 2)
approx_min_coarse = np.min(y_coarse)

print(f"Approximate minimum with 500 points: {approx_min_fine:.4f}")
print(f"Approximate minimum with 25 points: {approx_min_coarse:.4f}")
```

**E.8** Using logical (boolean) indexing, plot the function again but only the negative values.

```python
# Create boolean for negative values
neg = y < 0
x_neg = x[neg]
y_neg = y[neg]

plt.figure(figsize=(8, 5))
plt.plot(x_neg, y_neg, 'r.', markersize=4, label='Negative y values')
plt.xlabel('x')
plt.ylabel('y')
plt.title(r'Negative Values of $y=(x+4)(x-3)(x+1)^2$')
plt.axhline(0, color='black', linestyle='--', linewidth=0.8)
plt.grid(True)
plt.legend()
plt.show()
```
