**L.1** Calculate the 'factorial' of a number. #阶乘 in chinese

```python
# 5-->1, use for/while loop
n = 5
prod=1
temp=n
while temp>0:
    prod=prod*temp
    temp=temp-1
    print(prod)
    print(temp)
```

```python
n = 5
prod = 1

for temp in range(n, 0, -1):
    prod = prod * temp
    print(prod)
    print(temp)
```

**E.2** What is the difference between np.linspace and np.arange? Explain and show how you would use them to plot $y=x^3$ from -10 to 10

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Using arange
x1 = np.arange(-10, 10.1, 0.1)
y1 = x1**3

plt.plot(x1, y1)
plt.xlabel('x')
plt.ylabel('y')
plt.title('y = x^3 using np.arange')
plt.show()
```

```python
# Using linspace
x2 = np.linspace(-10, 10, 201)
y2 = x2**3

plt.plot(x2, y2)
plt.xlabel('x')
plt.ylabel('y')
plt.title('y = x^3 using np.linspace')
plt.show()
```

np.arange() creates values using a specified step size, while np.linspace() creates a specified number of evenly spaced values between two endpoints.
For example, np.arange(-10, 10.1, 0.1) uses a step size of 0.1, while np.linspace(-10, 10, 201) creates 201 evenly spaced points from -10 to 10.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Random numbers

# Set the random seed so the same random numbers can be reproduced
np.random.seed(123)

# Generate a random decimal number between 0 and 1
print(np.random.rand())

# Generate a random integer from 1 to 6
dice = np.random.randint(1, 7)
print(dice)
```

A random walk means that the next position depends on a random event.

```python
np.random.seed(123)

random_walk = [0]

for x in range(100):

    step = random_walk[-1]

    dice = np.random.randint(1, 7)

    if dice <= 2:
        step = max(0, step - 1)

    elif dice <= 5:
        step = step + 1

    else:
        step = step + np.random.randint(1, 7)

    random_walk.append(step)

print(random_walk)
```

A random walk changes step by step based on random events.

The values can be stored in a list and then plotted to see how the
position changes over time.

```python
plt.plot(random_walk)

plt.xlabel("Step")
plt.ylabel("Position")
plt.title("Random Walk")

plt.show()
```

Simulating many random walks

```python
np.random.seed(123)

all_walks = []

for i in range(500):

    random_walk = [0]

    for x in range(100):

        step = random_walk[-1]
        dice = np.random.randint(1, 7)

        if dice <= 2:
            step = max(0, step - 1)

        elif dice <= 5:
            step = step + 1

        else:
            step = step + np.random.randint(1, 7)

        random_walk.append(step)

    all_walks.append(random_walk)
```

Convert a list to a NumPy array

```python
np_aw = np.array(all_walks)

print(np_aw.shape)
```

Transpose an array

```python
np_aw_t = np.transpose(np_aw)

print(np_aw_t.shape)
```

Plot many random walks & Look at the final positions

```python
plt.plot(np_aw_t)

plt.xlabel("Step")
plt.ylabel("Position")
plt.title("500 Random Walks")

plt.show()
```

```python
ends = np_aw[:, -1]

print(ends)
```

Histogram

A histogram shows the distribution of simulation results.

Instead of looking at one random walk, we can look at hundreds of
simulations to estimate how likely different outcomes are.

```python
plt.hist(ends, bins=20)

plt.xlabel("Final Position")
plt.ylabel("Frequency")
plt.title("Distribution of Final Positions")

plt.show()
```

Estimate probability

```python
probability = np.mean(ends >= 60)

print(probability)
```
