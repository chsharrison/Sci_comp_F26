**W.2** Pandas filtering and sorting practice with Chipotle data

Import the dataset from this address https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv
Assign it to a variable called chipo.
1. How many products cost more than $10.00?
2. What is the price of each item?
    1. Print a data frame with only two columns item_name and item_price
    2. Sort by the name of the item
1. What was the quantity of the most expensive item ordered?
1. How many times did someone order more than one Canned Soda?


```python
# Read the supplied dataset, using a local copy when available.
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
data_file = Path("data/chipotle.tsv")
if data_file.exists():
    chipo = pd.read_csv(data_file, sep="\t")
else:
    chipo = pd.read_csv(url, sep="\t")
    data_file.parent.mkdir(parents=True, exist_ok=True)
    chipo.to_csv(data_file, sep="\t", index=False)

# Convert the recorded line totals from currency strings to numbers.
chipo["item_price"] = (
    chipo["item_price"].str.replace("$", "", regex=False).str.strip().astype(float)
)

# 1. Count records with a line total greater than $10.
products_over_10 = chipo.loc[chipo["item_price"] > 10]
print(f"1. Records with item_price > $10.00: {len(products_over_10)}")

# Distinguish product types from records containing multiple units.
unit_prices = chipo["item_price"] / chipo["quantity"]
expensive_item_names = chipo.loc[unit_prices > 10, "item_name"].nunique()
print(f"   Distinct item names with unit price > $10.00: {expensive_item_names}")

# 2. Show the two requested columns, sorted alphabetically by item name.
# Single-unit purchases give item prices without quantity effects.
# Keep different prices for the same name because customizations can differ.
item_prices = (
    chipo.loc[chipo["quantity"] == 1, ["item_name", "item_price"]]
    .drop_duplicates()
    .sort_values(["item_name", "item_price"])
    .reset_index(drop=True)
)
print("\n2. Item prices (single-unit purchases, sorted by item name):")
print(item_prices.to_string(index=False, float_format="${:.2f}".format))

# 3. Find the largest recorded line total and report its quantity.
most_expensive = chipo.sort_values("item_price", ascending=False).iloc[0]
print(f"\n3. Quantity in the most expensive record: {most_expensive['quantity']}")
print(f"   Item: {most_expensive['item_name']}; total: ${most_expensive['item_price']:.2f}")

# 4. Count matching occurrences, rather than summing the quantities.
multiple_canned_soda = chipo.loc[
    (chipo["item_name"] == "Canned Soda") & (chipo["quantity"] > 1)
]
print(f"\n4. Times more than one Canned Soda was ordered: {len(multiple_canned_soda)}")
```
