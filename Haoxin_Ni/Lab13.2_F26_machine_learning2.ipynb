**W.4 Pandas Practice con**

Compute the minimum, 25th percentile, median, 75th, and maximum of the following series:

ser = pd.Series(np.random.normal(10, 5, 25))

What kind of distribution is this?


```python
ser = pd.Series(np.random.normal(10, 5, 25))
summary = pd.Series({
    'Minimum': ser.min(),
    '25th percentile': ser.quantile(0.25),
    'Median': ser.median(),
    '75th percentile': ser.quantile(0.75),
    'Maximum': ser.max(),
})

display(summary)
print('The 25 values are sampled from a normal distribution with mean 10 and standard deviation 5.')
```

### scikit-learn cheat sheet

- **Classification metrics.** A confusion matrix counts correct and incorrect predictions by class. Precision measures how many predicted positives are correct. Recall measures how many actual positives are found. F1 is the harmonic mean of precision and recall. ROC AUC uses prediction scores, such as positive-class probabilities.
- **Tuning.** GridSearchCV tests every listed combination. RandomizedSearchCV samples combinations. Tune with cross-validation on the training set, then evaluate the selected model on the held-out test set.
- **Preprocessing.** Use pd.get_dummies() to encode categories, SimpleImputer to fill missing values, and StandardScaler to center and scale numeric features. A Pipeline fits preprocessing separately within each training fold, which prevents data leakage.

~~~python
from sklearn.datasets import load_iris
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, roc_auc_score
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

features, labels = load_iris(return_X_y=True)
features = features[labels != 0]
target = (labels[labels != 0] == 2).astype(int)
features[::10, 0] = float('nan')  # Add missing values for this example.
X_train, X_test, y_train, y_test = train_test_split(
    features, target, test_size=0.25, stratify=target, random_state=42
)
pipeline = Pipeline([
    ('imputer', SimpleImputer()),
    ('scaler', StandardScaler()),
    ('logreg', LogisticRegression()),
])
search = GridSearchCV(pipeline, {'logreg__C': [0.1, 1, 10]}, cv=5, scoring='roc_auc')
search.fit(X_train, y_train)
print(search.best_params_)
print(classification_report(y_test, search.predict(X_test)))
print('Test ROC AUC', roc_auc_score(y_test, search.predict_proba(X_test)[:, 1]))
~~~
