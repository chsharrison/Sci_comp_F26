```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    'ibtracs.NA.list.v04r01.csv',
    skiprows=[1],
    low_memory=False
)
```

```python
ida = df[
    (df['SEASON'] == 2021) &
    (df['NAME'].str.strip() == 'IDA')
].copy()
```

```python
ida = ida[
    ['ISO_TIME', 'USA_WIND', 'USA_PRES']
].copy()

ida.columns = ['time', 'wind_kt', 'pressure_hPa']

ida['time'] = pd.to_datetime(ida['time'])
ida['wind_kt'] = pd.to_numeric(ida['wind_kt'], errors='coerce')
ida['pressure_hPa'] = pd.to_numeric(
    ida['pressure_hPa'],
    errors='coerce'
)

ida.head()
```

```python
# Function

def analyze_intensity(data):
    """
    Prepare 6-hourly tropical cyclone data and calculate
    24-hour intensity changes.
    """

    df = data.copy()

    # Keep only standard 6-hourly observations
    df = df[
        (df['time'].dt.hour.isin([0, 6, 12, 18])) &
        (df['time'].dt.minute == 0)
    ].copy()

    # Sort observations by time
    df = df.sort_values('time').reset_index(drop=True)

    # Calculate 24-hour changes
    df['wind_change_24h'] = (
        df['wind_kt'] - df['wind_kt'].shift(4)
    )

    df['pressure_change_24h'] = (
        df['pressure_hPa'] - df['pressure_hPa'].shift(4)
    )

    # Identify rapid intensification
    df['RI'] = df['wind_change_24h'] >= 30

    return df
```

```python
ida_analysis = analyze_intensity(ida)

ida_analysis.head(10)
```

```python
ida_analysis[
    ida_analysis['RI']
][
    ['time', 'wind_kt',
     'wind_change_24h',
     'pressure_change_24h']
]
```

```python
# Plot Function

def plot_intensity(data, storm_name):
    
    """
    Plot maximum wind speed and minimum central pressure,
    and highlight rapid intensification points.
    """

    fig, ax1 = plt.subplots(figsize=(10, 5))

    # Wind speed
    ax1.plot(
        data['time'],
        data['wind_kt'],
        marker='o',
        linewidth=2,
        label='Maximum Wind'
    )

    ax1.set_xlabel('Date', fontsize=15, fontweight='bold')
    ax1.set_ylabel('Maximum Wind (kt)', fontsize=15, fontweight='bold')

    # Second y-axis for pressure
    ax2 = ax1.twinx()

    ax2.plot(
        data['time'],
        data['pressure_hPa'],
        linestyle='--',
        linewidth=2,
        color='darkorange',
        label='Minimum Pressure'
    )

    ax2.set_ylabel('Minimum Pressure (hPa)', fontsize=15, fontweight='bold')

    # Highlight rapid intensification points
    ri = data[data['RI']]

    ax1.scatter(
        ri['time'],
        ri['wind_kt'],
        s=90,
        marker='*',
        label='RI detected (24-h ΔV ≥ 30 kt)',
        zorder=5
    )

    ax1.set_title(
        storm_name + ' Intensity Evolution',
        fontsize=19,
        fontweight='bold',
        pad=15
    )

    ax1.spines['top'].set_visible(False)
    ax2.spines['top'].set_visible(False)

    lines1, labels1 = ax1.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()

    legend = ax1.legend(
        lines1 + lines2,
        labels1 + labels2,
        loc='center right',
        fontsize=12,
        frameon=True
    )

    ax1.tick_params(axis='both', labelsize=12)
    ax2.tick_params(axis='y', labelsize=12)

    for label in ax1.get_xticklabels():
        label.set_fontweight('bold')

    for label in ax1.get_yticklabels():
        label.set_fontweight('bold')

    for label in ax2.get_yticklabels():
        label.set_fontweight('bold')

    fig.autofmt_xdate()
    plt.tight_layout()
    plt.show()
```

```python
plot_intensity(ida_analysis, 'Hurricane Ida (2021)')
```
