**W.1** Write a function that returns a list of the first n powers of 2 (2^0, 2^1, 2^2, 2^3...2^n).

```python
def powers_of_two(n):
    powers = []
    for i in range(n + 1):
        powers.append(2 ** i)
    return powers

powers_of_two(5)
```

**W.2** Using the function above, plot the first 100 powers of 2

```python
# The first 100 powers are 2**0 through 2**99.
powers = powers_of_two(99)

plt.plot(range(100), powers)
plt.xlabel("Exponent")
plt.ylabel("Power of 2")
plt.title("First 100 Powers of 2")
plt.show()
```

### Notes: DataCamp Chapters 1-2

- Descriptive statistics summarize data; inferential statistics use a sample to learn about a population.
- The mean is the average, and the median is the middle value. The median is less affected by outliers.
- Variance and standard deviation measure spread. Use `ddof=1` for sample variance or standard deviation.
- IQR = Q3 - Q1. Values below Q1 - 1.5 * IQR or above Q3 + 1.5 * IQR may be outliers.
- `.value_counts()` counts categories, and `.sample()` takes a random sample. `replace=True` allows repeats; a random seed makes results reproducible.
- `uniform.cdf()` gives cumulative probabilities for a uniform distribution. Its `loc` is the minimum and `scale` is the range width.
- A binomial distribution counts successes in independent trials with the same success probability. `binom.pmf()` gives P(X = k), `binom.cdf()` gives P(X <= k), and the expected value is n * p.


```python
import numpy as np
import pandas as pd
from scipy.stats import uniform, binom

# Simple summary statistics
x = pd.Series([2, 4, 6, 8, 10])
print("Mean and median:", np.mean(x), np.median(x))
print("Sample standard deviation:", np.std(x, ddof=1))
q1, q3 = np.quantile(x, [0.25, 0.75])
print("IQR:", q3 - q1)

# Reproducible sampling with replacement
np.random.seed(24)
print("Sample:", x.sample(3, replace=True).tolist())

# A uniform wait time from 0 to 30 minutes
print("P(wait < 5):", uniform.cdf(5, loc=0, scale=30))

# Three independent deals, each with a 30% chance of success
print("P(exactly 1 success):", binom.pmf(1, 3, 0.3))
print("Expected successes:", 3 * 0.3)
```
