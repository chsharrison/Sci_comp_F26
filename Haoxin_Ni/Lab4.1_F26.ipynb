```python
import numpy as np

data = np.loadtxt('populations.txt')

years = data[:, 0]
populations = data[:, 1:]
species = np.array(['hare', 'lynx', 'carrot'])
```

```python
# 1. Mean and std of the populations of each species for the period
mean_pop = populations.mean(axis=0)
std_pop = populations.std(axis=0)

print("1. Mean and standard deviation:")
print("hare mean:", mean_pop[0], "std:", std_pop[0])
print("lynx mean:", mean_pop[1], "std:", std_pop[1])
print("carrot mean:", mean_pop[2], "std:", std_pop[2])
```

```python
# 2. Which year each species had the largest population
max_years_idx = np.argmax(populations, axis=0)

print("2. Max population year:")
print("hare max year:", years[max_years_idx[0]])
print("lynx max year:", years[max_years_idx[1]])
print("carrot max year:", years[max_years_idx[2]])
```

```python
# 3. Which species had the largest population each year
species = np.array(['hare', 'lynx', 'carrot'])
max_species_idx = np.argmax(populations, axis=1)

print("3. Max species each year:")
for i in range(len(years)):
    print(years[i], species[max_species_idx[i]])
```

```python
x = 10

print(x > 5)      # True
print(x < 5)      # False
print(x == 10)    # True
print(x != 10)    # False
print(x >= 10)    # True
print(x <= 10)    # True
```

2. Boolean operators

```python
temperature = 30
rain = False

print(temperature > 25 and rain == False)
print(temperature > 35 or rain == True)
print(not rain)
```

3. if, elif, else

```python
temperature = 28

if temperature > 30:
    print("It is hot.")
elif temperature > 20:
    print("It is warm.")
else:
    print("It is cool.")
```

4. NumPy boolean filtering

```python
import numpy as np

temps = np.array([18, 22, 27, 31, 35])

# Check which temperatures are above 30
print(temps > 30)

# Select only temperatures above 30
hot_temps = temps[temps > 30]

print(hot_temps)
```

```python
# Select temperatures between 20 and 30
comfortable = temps[np.logical_and(temps >= 20, temps <= 30)] # np.logical_and()

print(comfortable)
```

5. Pandas filtering

```python
import pandas as pd

weather = pd.DataFrame({
    "city": ["Baton Rouge", "New Orleans", "Houston"],
    "temperature": [32, 31, 34],
    "rain": [True, False, False]
})

# Select cities warmer than 32 degrees
warm = weather[weather["temperature"] > 32]

print(warm)
```
