### Notes: Statistics and HPC

- A normal distribution is symmetric and is described by its mean and standard deviation. `norm.cdf()` gives cumulative probabilities; `norm.ppf()` finds percentiles.
- With independent random observations of finite variance, the distribution of sample means becomes approximately normal as the sample size grows (CLT).
- Poisson models event counts in a fixed interval. Exponential models waiting times; its `scale` is the mean waiting time, or 1/rate. A t-distribution has heavier tails than a normal distribution.
- `.corr()` measures a linear relationship from -1 to 1. Always look at a scatterplot; a low correlation can hide a nonlinear relationship. `np.log()` can help with skewed positive data.
- Correlation does not prove causation. Random assignment and comparable control groups help reduce bias; observational studies can have confounders.
- Longitudinal studies follow the same subjects over time; cross-sectional studies take a snapshot.
- Open OnDemand gives browser access to HPC files, jobs, and Jupyter. Check the allocation and requested resources before starting a job.


```python
import numpy as np
import pandas as pd
from scipy.stats import norm, poisson, expon

# Simple probability examples
print("P(X < 7500):", norm.cdf(7500, loc=5000, scale=2000))
print("25th percentile:", norm.ppf(0.25, loc=5000, scale=2000))
print("P(5 events):", poisson.pmf(5, mu=4))
print("P(wait < 1 hour):", expon.cdf(1, scale=2.5))

# Repeat sampling and calculate sample means
np.random.seed(104)
x = pd.Series([1, 2, 3, 4, 5, 6])
sample_means = []
for i in range(100):
    sample = x.sample(20, replace=True)
    sample_means.append(sample.mean())
print("Mean of sample means:", np.mean(sample_means))

# Pearson correlation measures linear association
y = pd.Series([2, 4, 5, 8, 9, 12])
print("Correlation:", x.corr(y))
```
